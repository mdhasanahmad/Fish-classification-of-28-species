```python
# ================================================================
# ✅ Rathi et al. (Fish Disease CNN, Kaggle-Ready, 4-Channel Input)
# ================================================================

import os

# -----------------------------
# Fix 1: Protobuf compatibility monkeypatch
# -----------------------------
try:
    from google.protobuf import message_factory
    if not hasattr(message_factory.MessageFactory, "GetPrototype"):
        message_factory.MessageFactory.GetPrototype = (
            lambda self, descriptor: self.GetMessageClass(descriptor)
        )
        print("[INFO] Patched google.protobuf.message_factory.MessageFactory.GetPrototype -> GetMessageClass")
except Exception as e:
    print("[WARN] Could not apply protobuf monkeypatch:", e)

# -----------------------------
# Fix 2: Suppress noisy TF logs
# -----------------------------
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
try:
    import tensorflow as tf
    tf.get_logger().setLevel("ERROR")
except Exception as e:
    print("[WARN] TensorFlow import issue:", e)

# -----------------------------
# Other imports
# -----------------------------
import cv2
import numpy as np
import joblib
import matplotlib.pyplot as plt
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score, cohen_kappa_score, classification_report,
    confusion_matrix, ConfusionMatrixDisplay
)
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D, GlobalAveragePooling2D,
    Dense, Dropout
)
from tensorflow.keras.optimizers import Adam

# ================================================================
# CONFIGURATION
# ================================================================
DATA_DIR = "/kaggle/input/fresh-water-dataset/Original Image"
IMG_SIZE = 100
TEST_SPLIT = 0.20
RANDOM_STATE = 42
BATCH_SIZE = 32
EPOCHS = 20
VERBOSE = 1

# ================================================================
# DATASET UTILITIES
# ================================================================
def list_class_folders(data_dir):
    """Return sorted list of class folder names (only directories)."""
    if not os.path.exists(data_dir):
        raise FileNotFoundError(f"DATA_DIR not found: {data_dir}")
    classes = []
    with os.scandir(data_dir) as it:
        for entry in it:
            if entry.is_dir():
                classes.append(entry.name)
    return sorted(classes)

# ================================================================
# IMAGE PREPROCESSING
# ================================================================
def preprocess_image(img_path, img_size=IMG_SIZE):
    """
    Read image -> resize -> gray -> blur -> Otsu -> erosion -> dilation
    -> stack RGB(normalized) + mask into 4-channel float32 array [0,1].
    """
    img = cv2.imread(img_path)
    if img is None:
        raise ValueError("cv2.imread returned None for: " + str(img_path))
    img = cv2.resize(img, (img_size, img_size))
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(gray, (5, 5), 0)
    _, binary = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    kernel = np.ones((3, 3), np.uint8)
    eroded = cv2.erode(binary, kernel, iterations=1)
    dilated = cv2.dilate(eroded, kernel, iterations=1)

    rgb_norm = img_rgb.astype("float32") / 255.0
    mask = (dilated.astype("float32") / 255.0).reshape(img_size, img_size, 1)
    stacked = np.concatenate([rgb_norm, mask], axis=-1)
    return stacked

# ================================================================
# DATA LOADING
# ================================================================
def load_dataset(data_dir):
    """Load images and labels into arrays."""
    class_folders = list_class_folders(data_dir)
    if len(class_folders) == 0:
        raise ValueError("No class subfolders found in DATA_DIR.")

    total_files = 0
    for cls in class_folders:
        cls_path = os.path.join(data_dir, cls)
        total_files += len([
            f for f in os.listdir(cls_path)
            if f.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff'))
        ])
    print(f"[INFO] Found {len(class_folders)} classes and ~{total_files} image files (counted).")

    images, labels = [], []
    for cls in tqdm(class_folders, desc="Loading classes", unit="cls"):
        cls_path = os.path.join(data_dir, cls)
        for entry in os.scandir(cls_path):
            if entry.is_file() and entry.name.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff')):
                try:
                    arr = preprocess_image(os.path.join(cls_path, entry.name), IMG_SIZE)
                    images.append(arr)
                    labels.append(cls)
                except Exception:
                    continue

    X = np.array(images, dtype=np.float32)
    y = np.array(labels, dtype=object)
    return X, y, class_folders

# ================================================================
# FIXED CNN ARCHITECTURE (SAFE FOR 100x100)
# ================================================================
def build_rathi_cnn(input_shape, num_classes, dropout_rate=0.8):
    """
    Fixed CNN that avoids shape collapse and follows Rathi et al. logic.
    """
    model = Sequential()
    model.add(Input(shape=input_shape))

    model.add(Conv2D(32, (3, 3), activation='relu', padding='same'))
    model.add(MaxPooling2D((2, 2)))  # 100 → 50

    model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    model.add(MaxPooling2D((2, 2)))  # 50 → 25

    model.add(Conv2D(32, (3, 3), activation='relu', padding='same'))
    model.add(MaxPooling2D((2, 2)))  # 25 → 12

    model.add(GlobalAveragePooling2D())
    model.add(Dense(256, activation='relu'))
    model.add(Dropout(dropout_rate))
    model.add(Dense(num_classes, activation='softmax'))

    model.compile(optimizer=Adam(1e-4),
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])
    return model

# ================================================================
# MAIN TRAINING + EVALUATION
# ================================================================
def main():
    print("[INFO] Loading dataset from:", DATA_DIR)
    X, y, class_folders = load_dataset(DATA_DIR)
    print(f"[INFO] Loaded dataset: X.shape={X.shape}, labels={len(np.unique(y))} unique classes")

    le = LabelEncoder()
    y_enc = le.fit_transform(y)
    y_cat = to_categorical(y_enc)

    X_train, X_test, y_train, y_test = train_test_split(
        X, y_cat, test_size=TEST_SPLIT,
        random_state=RANDOM_STATE, stratify=y_cat
    )
    print("[INFO] Train shape:", X_train.shape, "Test shape:", X_test.shape)

    model = build_rathi_cnn((IMG_SIZE, IMG_SIZE, 4), num_classes=len(class_folders))
    model.summary()

    history = model.fit(
        X_train, y_train,
        validation_data=(X_test, y_test),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=VERBOSE
    )

    # ================================================================
    # EVALUATION
    # ================================================================
    y_pred_probs = model.predict(X_test, verbose=0)
    y_pred = np.argmax(y_pred_probs, axis=1)
    y_true = np.argmax(y_test, axis=1)

    acc = accuracy_score(y_true, y_pred)
    kappa = cohen_kappa_score(y_true, y_pred)
    report = classification_report(y_true, y_pred, target_names=le.classes_, digits=4)
    cm = confusion_matrix(y_true, y_pred)

    print("===============================================================")
    print(f"✅ Accuracy: {acc*100:.2f}%")
    print(f"✅ Cohen's Kappa: {kappa:.4f}")
    print("===============================================================")
    print("Classification Report:\n", report)

    ConfusionMatrixDisplay.from_predictions(
        y_true, y_pred,
        display_labels=le.classes_,
        xticks_rotation=90, cmap='Blues'
    )
    plt.title("Confusion Matrix – Rathi CNN (Fixed)")
    plt.show()

    # ================================================================
    # SAVE MODEL + ENCODER
    # ================================================================
    model.save("rathi_underwater_cnn_fixed.h5")
    joblib.dump(le, "label_encoder.joblib")
    print("[INFO] Saved model -> rathi_underwater_cnn_fixed.h5 and label encoder -> label_encoder.joblib")

    # ================================================================
    # PLOT TRAINING CURVES
    # ================================================================
    try:
        plt.figure(figsize=(10, 4))
        plt.subplot(1, 2, 1)
        plt.plot(history.history['loss'], label='train_loss')
        plt.plot(history.history['val_loss'], label='val_loss')
        plt.title('Loss')
        plt.legend()

        plt.subplot(1, 2, 2)
        plt.plot(history.history['accuracy'], label='train_acc')
        plt.plot(history.history['val_accuracy'], label='val_acc')
        plt.title('Accuracy')
        plt.legend()
        plt.show()
    except Exception:
        pass

# ================================================================
# RUN
# ================================================================
if __name__ == "__main__":
    main()
```
