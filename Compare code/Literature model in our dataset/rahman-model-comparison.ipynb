```python
# resnet50_svc_optimized.py
# ResNet50 (fine-tuned) -> Feature extraction -> PCA -> SVC (tuned)
# Expected: significant accuracy improvements vs. frozen ImageNet features

import os
import numpy as np
import joblib
import matplotlib.pyplot as plt
from pathlib import Path

# sklearn
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.metrics import (
    classification_report, confusion_matrix, accuracy_score, cohen_kappa_score
)
from sklearn.model_selection import train_test_split

# TensorFlow / Keras
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
import tensorflow as tf
tf.get_logger().setLevel('ERROR')

from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.optimizers import Adam

# -----------------------------
# USER PARAMETERS - adjust these
# -----------------------------
DATA_DIR = "/kaggle/input/fresh-water-dataset/Original Image"  # <-- update if needed
IMG_SIZE = (320, 320)   # higher res (was 224)
BATCH_SIZE = 16
RANDOM_STATE = 42
FT_EPOCHS = 6          # fine-tune epochs; increase if you have time/GPU
FT_BATCH_SIZE = BATCH_SIZE
PCA_VARIANCE = 0.95     # keep 95% variance
TEST_SIZE = 0.20
SAVE_DIR = "artifacts"
os.makedirs(SAVE_DIR, exist_ok=True)
VERBOSE = 1

# -----------------------------
# 1) Data generators (augmentation for training)
# -----------------------------
train_aug = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=20,
    width_shift_range=0.08,
    height_shift_range=0.08,
    shear_range=0.12,
    zoom_range=0.12,
    horizontal_flip=True,
    fill_mode="nearest",
    validation_split=0.15  # keep a small validation split for fine-tuning
)

val_aug = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    validation_split=0.15
)

# Use subset split for fine-tuning training/validation
train_gen = train_aug.flow_from_directory(
    DATA_DIR,
    target_size=IMG_SIZE,
    batch_size=FT_BATCH_SIZE,
    class_mode="categorical",
    subset="training",
    shuffle=True,
    seed=RANDOM_STATE
)

val_gen = val_aug.flow_from_directory(
    DATA_DIR,
    target_size=IMG_SIZE,
    batch_size=FT_BATCH_SIZE,
    class_mode="categorical",
    subset="validation",
    shuffle=True,
    seed=RANDOM_STATE
)

NUM_CLASSES = train_gen.num_classes
CLASS_NAMES = list(train_gen.class_indices.keys())
print(f"[INFO] Classes ({NUM_CLASSES}): {CLASS_NAMES}")

# -----------------------------
# 2) Build ResNet50 base + small classifier head for fine-tuning
# -----------------------------
def build_finetune_model(img_size=IMG_SIZE, num_classes=NUM_CLASSES, unfreeze_last_n=30):
    base = ResNet50(weights="imagenet", include_top=False,
                    input_shape=(img_size[0], img_size[1], 3))
    # Freeze all but last `unfreeze_last_n` layers
    if unfreeze_last_n is not None and unfreeze_last_n > 0:
        for layer in base.layers[:-unfreeze_last_n]:
            layer.trainable = False
        for layer in base.layers[-unfreeze_last_n:]:
            layer.trainable = True
    else:
        for layer in base.layers:
            layer.trainable = False

    x = GlobalAveragePooling2D()(base.output)
    x = Dropout(0.3)(x)
    out = Dense(num_classes, activation="softmax")(x)
    model = Model(inputs=base.input, outputs=out)
    return model, base

print("[INFO] Building model for fine-tuning...")
ft_model, base_model = build_finetune_model(unfreeze_last_n=30)
ft_model.compile(optimizer=Adam(learning_rate=1e-4),
                 loss="categorical_crossentropy", metrics=["accuracy"])
ft_model.summary()

# -----------------------------
# 3) Fine-tune (train head + last layers)
# -----------------------------
# Use a small number of steps per epoch to keep short by default; adjust if needed.
steps_per_epoch = max(1, train_gen.samples // FT_BATCH_SIZE)
validation_steps = max(1, val_gen.samples // FT_BATCH_SIZE)

print(f"[INFO] Fine-tuning for {FT_EPOCHS} epochs...")
history = ft_model.fit(
    train_gen,
    steps_per_epoch=steps_per_epoch,
    validation_data=val_gen,
    validation_steps=validation_steps,
    epochs=FT_EPOCHS,
    verbose=VERBOSE
)

# Optionally: save the fine-tuned model weights
ft_model.save(Path(SAVE_DIR) / "resnet50_finetuned.h5")
print("[INFO] Fine-tuned model saved.")

# -----------------------------
# 4) Create feature extractor that outputs GAP features from (fine-tuned) base
# -----------------------------
feature_extractor = Model(inputs=base_model.input,
                          outputs=GlobalAveragePooling2D()(base_model.output))
print("[INFO] Feature extractor ready (using fine-tuned ResNet50 base).")

# -----------------------------
# 5) Extract features for whole dataset (no augmentation, shuffle=False)
# -----------------------------
extract_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)
extract_gen = extract_datagen.flow_from_directory(
    DATA_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

print(f"[INFO] Extracting features for {extract_gen.samples} images ...")
features = feature_extractor.predict(extract_gen, verbose=VERBOSE)
# labels in the same order as extract_gen.filenames because shuffle=False
labels_idx = extract_gen.classes
class_names = list(extract_gen.class_indices.keys())
labels = np.array([class_names[i] for i in labels_idx])
print(f"[INFO] Extracted features shape: {features.shape}")

# -----------------------------
# 6) Encode labels, split, scale and PCA
# -----------------------------
le = LabelEncoder()
y_enc = le.fit_transform(labels)

X_train, X_test, y_train, y_test = train_test_split(
    features, y_enc, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=y_enc
)

scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

# PCA to reduce dimensionality (removes redundancy -> SVC performs better)
pca = PCA(n_components=PCA_VARIANCE, svd_solver="full", random_state=RANDOM_STATE)
X_train_p = pca.fit_transform(X_train_s)
X_test_p = pca.transform(X_test_s)
print(f"[INFO] After PCA: train shape = {X_train_p.shape}, test shape = {X_test_p.shape}")

# -----------------------------
# 7) Train SVC (tuned)
# -----------------------------
# Recommended settings: C larger (10), class_weight='balanced', probability=True for better boundaries
svc = SVC(C=10.0, kernel='rbf', gamma='scale', class_weight='balanced', probability=True,
          random_state=RANDOM_STATE)
print("[INFO] Training SVC ...")
svc.fit(X_train_p, y_train)

# If you want to run GridSearchCV (slower), uncomment below:
# from sklearn.model_selection import GridSearchCV
# param_grid = {
#     'C': [1, 10, 50],
#     'gamma': ['scale', 0.01, 0.001],
#     'kernel': ['rbf']
# }
# grid = GridSearchCV(SVC(class_weight='balanced', probability=True, random_state=RANDOM_STATE),
#                     param_grid, cv=3, n_jobs=-1, verbose=2)
# grid.fit(X_train_p, y_train)
# svc = grid.best_estimator_
# print(f"[INFO] Best SVC from grid: {grid.best_params_}")

# -----------------------------
# 8) Evaluation
# -----------------------------
y_pred = svc.predict(X_test_p)

acc = accuracy_score(y_test, y_pred)
kappa = cohen_kappa_score(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=le.classes_)
cm = confusion_matrix(y_test, y_pred)

print("===============================================================")
print(f"✅ Accuracy: {acc*100:.4f}%")
print(f"✅ Cohen's Kappa: {kappa:.4f}")
print("===============================================================")
print("Classification Report:\n", report)

# Confusion matrix plot
try:
    from sklearn.metrics import ConfusionMatrixDisplay
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=le.classes_)
    disp.plot(xticks_rotation=90)
    plt.title("Confusion Matrix – ResNet50 (fine-tuned) + PCA + SVC")
    plt.tight_layout()
    plt.show()
except Exception as e:
    print("[WARN] Could not display confusion matrix plot:", e)

# -----------------------------
# 9) Save artifacts (svc, scaler, pca, label encoder, class names)
# -----------------------------
joblib.dump(svc, Path(SAVE_DIR) / "svc_model.joblib")
joblib.dump(scaler, Path(SAVE_DIR) / "scaler.joblib")
joblib.dump(pca, Path(SAVE_DIR) / "pca.joblib")
np.save(Path(SAVE_DIR) / "label_classes.npy", le.classes_)
print(f"[INFO] Saved svc, scaler, pca, and label classes to '{SAVE_DIR}/'")

# -----------------------------
# 10) Quick inference helper (example)
# -----------------------------
def predict_from_image_paths(image_paths):
    """
    image_paths: list of file paths
    Returns: list of (pred_label, class_probabilities)
    """
    import cv2
    imgs = []
    for p in image_paths:
        img = cv2.imread(str(p))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, IMG_SIZE)
        img = preprocess_input(img.astype("float32"))
        imgs.append(img)
    X = np.array(imgs)
    feats = feature_extractor.predict(X, verbose=0)
    feats_s = scaler.transform(feats)
    feats_p = pca.transform(feats_s)
    probs = svc.predict_proba(feats_p)
    preds = svc.predict(feats_p)
    labels_out = [le.inverse_transform([p])[0] for p in preds]
    return list(zip(labels_out, probs))

# Example usage (uncomment and replace with actual paths):
# print(predict_from_image_paths(["/path/to/img1.jpg", "/path/to/img2.jpg"]))

print("[INFO] Done.")
```
