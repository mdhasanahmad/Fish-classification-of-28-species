```python
# ============================================================
# MOBILENET CLASSIFICATION
# ACTUAL GPU MEMORY + FLOPs + PARAMETERS + INFERENCE TIME
# TRAIN / VALIDATION / TEST ACCURACY WITH 95% CI
# ============================================================

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os
import glob
import time
import gc

from collections import Counter

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_curve,
    auc,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from sklearn.preprocessing import LabelBinarizer

from tensorflow.keras.preprocessing.image import ImageDataGenerator

from tensorflow.keras.applications.mobilenet import (
    MobileNet,
    preprocess_input
)

from tensorflow.keras import Model, layers

from tensorflow.keras.utils import to_categorical


# ============================================================
# 1. GPU CONFIGURATION
# ============================================================

print("=" * 80)
print("SYSTEM INFORMATION")
print("=" * 80)

gpus = tf.config.list_physical_devices("GPU")

if gpus:

    print("GPU detected:")

    for gpu in gpus:
        print("   ", gpu)

    try:

        for gpu in gpus:
            tf.config.experimental.set_memory_growth(
                gpu,
                True
            )

    except RuntimeError as e:

        print(
            "Memory growth configuration:",
            e
        )

else:

    print(
        "WARNING: No GPU detected. "
        "Running on CPU."
    )

print(
    "TensorFlow version:",
    tf.__version__
)

print("=" * 80)


# ============================================================
# 2. DATASET PATH
# ============================================================

DATASET_PATH = (
    "/kaggle/input/datasets/mdhasanahmad/fresh-water-dataset/Original Image"
)


# ============================================================
# 3. LOAD DATA
# ============================================================

classes = sorted(
    [
        d for d in os.listdir(DATASET_PATH)
        if os.path.isdir(
            os.path.join(
                DATASET_PATH,
                d
            )
        )
    ]
)

NUM_CLASSES = len(classes)

image_paths = []
image_labels = []


for idx, class_name in enumerate(classes):

    class_folder = os.path.join(
        DATASET_PATH,
        class_name
    )

    files = []

    for ext in [
        "*.jpg",
        "*.jpeg",
        "*.png",
        "*.bmp",
        "*.JPG",
        "*.JPEG",
        "*.PNG"
    ]:

        files.extend(
            glob.glob(
                os.path.join(
                    class_folder,
                    ext
                )
            )
        )

    image_paths.extend(files)

    image_labels.extend(
        [idx] * len(files)
    )


image_paths = np.array(
    image_paths
)

image_labels = np.array(
    image_labels
)


print(
    f"\nTotal images found: "
    f"{len(image_paths)}"
)

print(
    f"Classes detected: "
    f"{classes}"
)


# ============================================================
# 4. CLASS DISTRIBUTION
# ============================================================

label_counts = Counter(
    image_labels
)

plt.figure(
    figsize=(10, 6)
)

plt.bar(
    [
        classes[i]
        for i in label_counts.keys()
    ],
    label_counts.values(),
    color="teal"
)

plt.title(
    "Number of Images per Class"
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Image Count"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# ============================================================
# 5. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# 70% TRAIN
# 10% VALIDATION
# 20% TEST

train_paths, temp_paths, train_labels, temp_labels = train_test_split(
    image_paths,
    image_labels,
    test_size=0.30,
    stratify=image_labels,
    random_state=42
)


val_paths, test_paths, val_labels, test_labels = train_test_split(
    temp_paths,
    temp_labels,
    test_size=2/3,
    stratify=temp_labels,
    random_state=42
)


print(
    "\nDataset split:"
)

print(
    f"Train      : {len(train_paths)}"
)

print(
    f"Validation : {len(val_paths)}"
)

print(
    f"Test       : {len(test_paths)}"
)


print(
    "\nDataset percentages:"
)

print(
    f"Train      : "
    f"{len(train_paths) / len(image_paths) * 100:.2f}%"
)

print(
    f"Validation : "
    f"{len(val_paths) / len(image_paths) * 100:.2f}%"
)

print(
    f"Test       : "
    f"{len(test_paths) / len(image_paths) * 100:.2f}%"
)


# ============================================================
# 6. IMAGE CONFIGURATION
# ============================================================

IMG_SIZE = (224, 224)

BATCH_SIZE = 32


# ============================================================
# 7. DATA GENERATOR
# ============================================================

def path_label_generator(
    paths,
    labels,
    batch_size=BATCH_SIZE,
    shuffle=True,
    augment=False
):

    datagen = ImageDataGenerator(

        preprocessing_function=preprocess_input,

        shear_range=(
            0.1 if augment else 0
        ),

        zoom_range=(
            0.2 if augment else 0
        ),

        horizontal_flip=augment
    )


    while True:

        if shuffle:

            indices = np.random.permutation(
                len(paths)
            )

        else:

            indices = np.arange(
                len(paths)
            )


        for start in range(
            0,
            len(paths),
            batch_size
        ):

            batch_idx = indices[
                start:start + batch_size
            ]

            batch_images = []
            batch_labels = []


            for i in batch_idx:

                img = tf.keras.preprocessing.image.load_img(
                    paths[i],
                    target_size=IMG_SIZE
                )

                img = tf.keras.preprocessing.image.img_to_array(
                    img
                )

                batch_images.append(
                    img
                )

                batch_labels.append(
                    labels[i]
                )


            batch_images = np.array(
                batch_images,
                dtype=np.float32
            )


            batch_labels = to_categorical(
                batch_labels,
                num_classes=NUM_CLASSES
            )


            batch_images = datagen.standardize(
                batch_images
            )


            yield (
                batch_images,
                batch_labels
            )


# ============================================================
# 8. CREATE GENERATORS
# ============================================================

train_generator = path_label_generator(
    train_paths,
    train_labels,
    shuffle=True,
    augment=True
)

val_generator = path_label_generator(
    val_paths,
    val_labels,
    shuffle=False,
    augment=False
)

test_generator = path_label_generator(
    test_paths,
    test_labels,
    shuffle=False,
    augment=False
)


# ============================================================
# 9. STEPS
# ============================================================

# IMPORTANT:
# ceil() ensures the final incomplete batch is included.

train_steps = int(
    np.ceil(
        len(train_paths) /
        BATCH_SIZE
    )
)

val_steps = int(
    np.ceil(
        len(val_paths) /
        BATCH_SIZE
    )
)

test_steps = int(
    np.ceil(
        len(test_paths) /
        BATCH_SIZE
    )
)


print(
    "\nSteps:"
)

print(
    f"Train steps      : {train_steps}"
)

print(
    f"Validation steps : {val_steps}"
)

print(
    f"Test steps       : {test_steps}"
)


# ============================================================
# 10. MODEL - MOBILENET
# ============================================================

print(
    "\nBuilding MobileNet..."
)


conv_base = MobileNet(
    include_top=False,
    weights="imagenet",
    input_shape=(
        *IMG_SIZE,
        3
    )
)


# Freeze convolutional base

for layer in conv_base.layers:

    layer.trainable = False


# ============================================================
# 11. CLASSIFICATION HEAD
# ============================================================

x = conv_base.output


x = layers.GlobalAveragePooling2D()(
    x
)


x = layers.Dense(
    128,
    activation="relu"
)(x)


x = layers.Dropout(
    0.5
)(x)


outputs = layers.Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)


model = Model(
    inputs=conv_base.input,
    outputs=outputs
)


# ============================================================
# 12. COMPILE
# ============================================================

model.compile(

    optimizer="adam",

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


# ============================================================
# 13. MODEL SUMMARY
# ============================================================

model.summary()


# ============================================================
# 14. TRAIN
# ============================================================

EPOCHS = 10


print(
    "\n" + "=" * 80
)

print(
    "STARTING MOBILENET TRAINING"
)

print(
    "=" * 80
)


history = model.fit(

    train_generator,

    steps_per_epoch=train_steps,

    validation_data=val_generator,

    validation_steps=val_steps,

    epochs=EPOCHS,

    verbose=1
)


# ============================================================
# 15. FINAL TRAIN / VALIDATION ACCURACY
# ============================================================

train_acc = history.history[
    "accuracy"
][-1]

val_acc = history.history[
    "val_accuracy"
][-1]


print(
    f"\nFinal Train Accuracy: "
    f"{train_acc:.4f} "
    f"({train_acc * 100:.2f}%)"
)

print(
    f"Final Validation Accuracy: "
    f"{val_acc:.4f} "
    f"({val_acc * 100:.2f}%)"
)


# ============================================================
# 16. PREDICTION HELPER
# ============================================================

def get_split_predictions(
    paths,
    labels,
    steps
):

    gen = path_label_generator(

        paths,

        labels,

        batch_size=BATCH_SIZE,

        shuffle=False,

        augment=False
    )


    preds = []
    trues = []


    for _ in range(steps):

        x_batch, y_batch = next(
            gen
        )


        p = model.predict(
            x_batch,
            verbose=0
        )


        preds.append(
            p
        )

        trues.append(
            y_batch
        )


    preds = np.vstack(
        preds
    )

    trues = np.vstack(
        trues
    )


    return (

        np.argmax(
            trues,
            axis=1
        ),

        np.argmax(
            preds,
            axis=1
        ),

        preds

    )


# ============================================================
# 17. GET PREDICTIONS
# ============================================================

print(
    "\nExtracting predictions..."
)


y_train, y_train_pred, y_train_prob = (
    get_split_predictions(
        train_paths,
        train_labels,
        train_steps
    )
)


y_val, y_val_pred, y_val_prob = (
    get_split_predictions(
        val_paths,
        val_labels,
        val_steps
    )
)


y_test, y_test_pred, y_test_prob = (
    get_split_predictions(
        test_paths,
        test_labels,
        test_steps
    )
)


# ============================================================
# 18. BOOTSTRAP METRICS
# ============================================================

def bootstrap_metrics(
    y_true,
    y_pred,
    y_prob,
    n_bootstrap=1000
):

    rng = np.random.RandomState(
        42
    )


    acc = []
    prec = []
    rec = []
    f1 = []
    aucs = []


    for _ in range(
        n_bootstrap
    ):

        idx = rng.randint(
            0,
            len(y_true),
            len(y_true)
        )


        y_t = y_true[idx]

        y_p = y_pred[idx]

        y_pr = y_prob[idx]


        # Require all classes for multiclass AUC

        if len(
            np.unique(y_t)
        ) < NUM_CLASSES:

            continue


        acc.append(
            accuracy_score(
                y_t,
                y_p
            )
        )


        prec.append(
            precision_score(
                y_t,
                y_p,
                average="macro",
                zero_division=0
            )
        )


        rec.append(
            recall_score(
                y_t,
                y_p,
                average="macro",
                zero_division=0
            )
        )


        f1.append(
            f1_score(
                y_t,
                y_p,
                average="macro",
                zero_division=0
            )
        )


        try:

            aucs.append(

                roc_auc_score(

                    tf.keras.utils.to_categorical(
                        y_t,
                        num_classes=NUM_CLASSES
                    ),

                    y_pr,

                    multi_class="ovr",

                    average="macro"
                )

            )

        except Exception:

            continue


    def ci(values):

        values = np.asarray(
            values
        )

        return (

            np.mean(values),

            np.percentile(
                values,
                2.5
            ),

            np.percentile(
                values,
                97.5
            )
        )


    return (

        ci(acc),

        ci(prec),

        ci(rec),

        ci(f1),

        ci(aucs)

    )


# ============================================================
# 19. PRINT RESULTS
# ============================================================

def print_results(
    dataset_name,
    y_true,
    y_pred,
    y_prob
):

    (
        acc_ci,
        prec_ci,
        rec_ci,
        f1_ci,
        auc_ci
    ) = bootstrap_metrics(

        y_true,
        y_pred,
        y_prob
    )


    print(
        f"\n{'=' * 70}"
    )

    print(
        f"{dataset_name.upper()} RESULTS (95% CI)"
    )

    print(
        f"{'=' * 70}"
    )


    print(

        f"Accuracy: "
        f"{acc_ci[0]:.4f} "
        f"(95% CI: "
        f"{acc_ci[1]:.4f} – "
        f"{acc_ci[2]:.4f})"

    )


    print(

        f"Precision (Macro): "
        f"{prec_ci[0]:.4f} "
        f"(95% CI: "
        f"{prec_ci[1]:.4f} – "
        f"{prec_ci[2]:.4f})"

    )


    print(

        f"Recall (Macro): "
        f"{rec_ci[0]:.4f} "
        f"(95% CI: "
        f"{rec_ci[1]:.4f} – "
        f"{rec_ci[2]:.4f})"

    )


    print(

        f"F1-score (Macro): "
        f"{f1_ci[0]:.4f} "
        f"(95% CI: "
        f"{f1_ci[1]:.4f} – "
        f"{f1_ci[2]:.4f})"

    )


    print(

        f"AUC (OVR Macro): "
        f"{auc_ci[0]:.4f} "
        f"(95% CI: "
        f"{auc_ci[1]:.4f} – "
        f"{auc_ci[2]:.4f})"

    )


# ============================================================
# 20. FINAL METRICS
# ============================================================

print_results(
    "Train",
    y_train,
    y_train_pred,
    y_train_prob
)


print_results(
    "Validation",
    y_val,
    y_val_pred,
    y_val_prob
)


print_results(
    "Test",
    y_test,
    y_test_pred,
    y_test_prob
)


# ============================================================
# 21. ACTUAL GPU MEMORY FUNCTION
# ============================================================

def get_gpu_memory():

    gpus = tf.config.list_physical_devices(
        "GPU"
    )


    if not gpus:

        return np.nan, np.nan


    try:

        memory_info = (
            tf.config.experimental.get_memory_info(
                "GPU:0"
            )
        )


        current_mb = (

            memory_info["current"]
            /
            (1024 ** 2)

        )


        peak_mb = (

            memory_info["peak"]
            /
            (1024 ** 2)

        )


        return (
            current_mb,
            peak_mb
        )


    except Exception as e:

        print(
            "Unable to read TensorFlow GPU memory:",
            e
        )

        return (
            np.nan,
            np.nan
        )


# ============================================================
# 22. RESET GPU MEMORY STATISTICS
# ============================================================

if gpus:

    try:

        tf.config.experimental.reset_memory_stats(
            "GPU:0"
        )

        print(
            "\nGPU memory statistics reset."
        )

    except Exception as e:

        print(
            "Could not reset GPU memory statistics:",
            e
        )


# ============================================================
# 23. FRESH PERFORMANCE GENERATOR
# ============================================================

performance_test_gen = path_label_generator(

    test_paths,

    test_labels,

    batch_size=BATCH_SIZE,

    shuffle=False,

    augment=False
)


# ============================================================
# 24. WARM-UP
# ============================================================

print(
    "\nWarming up MobileNet..."
)


warmup_x, _ = next(
    performance_test_gen
)


for _ in range(5):

    output = model(
        warmup_x,
        training=False
    )

    # Force computation

    _ = tf.reduce_sum(
        output
    ).numpy()


# ============================================================
# 25. RESET MEMORY AFTER WARM-UP
# ============================================================

if gpus:

    try:

        tf.config.experimental.reset_memory_stats(
            "GPU:0"
        )

    except Exception:
        pass


# ============================================================
# 26. INFERENCE TIME MEASUREMENT
# ============================================================

inference_times = []


# Measure all test batches

num_measurement_batches = test_steps


print(
    f"\nMeasuring inference time "
    f"over {num_measurement_batches} batches..."
)


for _ in range(
    num_measurement_batches
):

    x_batch, _ = next(
        performance_test_gen
    )


    start_t = time.perf_counter()


    predictions = model(
        x_batch,
        training=False
    )


    # IMPORTANT:
    # Force TensorFlow/GPU execution
    # before stopping timer.

    _ = tf.reduce_sum(
        predictions
    ).numpy()


    end_t = time.perf_counter()


    batch_time_ms = (

        end_t - start_t

    ) * 1000.0


    actual_batch_size = (
        x_batch.shape[0]
    )


    per_image_time_ms = (

        batch_time_ms
        /
        actual_batch_size

    )


    inference_times.append(
        per_image_time_ms
    )


avg_inference_time_ms = np.mean(
    inference_times
)


# ============================================================
# 27. READ ACTUAL PEAK GPU MEMORY
# ============================================================

current_memory_mb, peak_memory_mb = (
    get_gpu_memory()
)


# ============================================================
# 28. MODEL PARAMETERS
# ============================================================

total_params = model.count_params()

total_params_m = (
    total_params / 1e6
)


# ============================================================
# 29. MOBILE NET FLOPs
# ============================================================

# Standard approximate FLOPs for
# MobileNet v1 at 224x224.

gflops = 1.15


# ============================================================
# 30. TEST ACCURACY
# ============================================================

test_accuracy_percentage = (

    accuracy_score(
        y_test,
        y_test_pred
    )

    * 100.0

)


# ============================================================
# 31. PERFORMANCE TABLE
# ============================================================

print(
    "\n" + "=" * 125
)

print(
    f"{'Performance & Efficiency Metrics Evaluation':^125}"
)

print(
    "=" * 125
)


header = (

    f"{'Method':<24} | "

    f"{'FLOPS (GFLOPS)':<16} | "

    f"{'Parameters (M)':<16} | "

    f"{'Memory Usage (MB)':<20} | "

    f"{'Inference Time (ms)':<21} | "

    f"{'Accuracy (%)':<13}"

)


print(
    header
)

print(
    "-" * 125
)


row = (

    f"{'MobileNet (Proposed)':<24} | "

    f"{gflops:<16.2f} | "

    f"{total_params_m:<16.2f} | "

    f"{peak_memory_mb:<20.2f} | "

    f"{avg_inference_time_ms:<21.2f} | "

    f"{test_accuracy_percentage:<13.2f}"

)


print(
    row
)

print(
    "=" * 125
)


# ============================================================
# 32. DETAILED PERFORMANCE INFORMATION
# ============================================================

print(
    "\n" + "=" * 70
)

print(
    "DETAILED PERFORMANCE INFORMATION"
)

print(
    "=" * 70
)


if not np.isnan(
    current_memory_mb
):

    print(

        f"Current GPU memory : "
        f"{current_memory_mb:.2f} MB"

    )

    print(

        f"Peak GPU memory    : "
        f"{peak_memory_mb:.2f} MB"

    )

else:

    print(
        "GPU memory          : "
        "Unavailable"
    )


print(

    f"Total parameters    : "
    f"{total_params:,}"

)


print(

    f"Parameters          : "
    f"{total_params_m:.2f} M"

)


print(

    f"FLOPs               : "
    f"{gflops:.2f} GFLOPs"

)


print(

    f"Inference time      : "
    f"{avg_inference_time_ms:.4f} ms/image"

)


print(

    f"Test accuracy       : "
    f"{test_accuracy_percentage:.2f}%"

)


print(
    "=" * 70
)


# ============================================================
# 33. CLASSIFICATION REPORT
# ============================================================

report = classification_report(

    y_test,

    y_test_pred,

    target_names=classes

)


print(
    "\nClassification Report:\n"
)

print(
    report
)


# ============================================================
# 34. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    y_test_pred

)


df_cm = pd.DataFrame(

    cm,

    index=classes,

    columns=classes

)


plt.figure(
    figsize=(8, 6)
)


sns.heatmap(

    df_cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    cbar=True,

    xticklabels=classes,

    yticklabels=classes

)


plt.title(
    "Confusion Matrix - Test Set"
)

plt.ylabel(
    "True Label"
)

plt.xlabel(
    "Predicted Label"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(
    rotation=0
)

plt.tight_layout()

plt.show()


# ============================================================
# 35. TRAINING CURVES
# ============================================================

acc = history.history[
    "accuracy"
]

val_acc = history.history[
    "val_accuracy"
]

loss = history.history[
    "loss"
]

val_loss = history.history[
    "val_loss"
]


plt.figure(
    figsize=(12, 5)
)


# Accuracy

plt.subplot(
    1,
    2,
    1
)


plt.plot(

    range(
        1,
        EPOCHS + 1
    ),

    acc,

    label="Training Accuracy",

    linewidth=2

)


plt.plot(

    range(
        1,
        EPOCHS + 1
    ),

    val_acc,

    label="Validation Accuracy",

    linewidth=2

)


plt.legend()

plt.title(
    "Accuracy per Epoch"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)


# Loss

plt.subplot(
    1,
    2,
    2
)


plt.plot(

    range(
        1,
        EPOCHS + 1
    ),

    loss,

    label="Training Loss",

    linewidth=2

)


plt.plot(

    range(
        1,
        EPOCHS + 1
    ),

    val_loss,

    label="Validation Loss",

    linewidth=2

)


plt.legend()

plt.title(
    "Loss per Epoch"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)


plt.tight_layout()

plt.show()


# ============================================================
# 36. MULTI-CLASS ROC CURVES
# ============================================================

lb = LabelBinarizer()


y_test_binarized = lb.fit_transform(
    y_test
)


# Binary correction

if (

    y_test_binarized.shape[1] == 1

    and

    NUM_CLASSES == 2

):

    y_test_binarized = np.hstack(

        [

            1 - y_test_binarized,

            y_test_binarized

        ]

    )


plt.figure(
    figsize=(9, 7)
)


for i in range(
    NUM_CLASSES
):

    fpr, tpr, _ = roc_curve(

        y_test_binarized[:, i],

        y_test_prob[:, i]

    )


    roc_auc = auc(
        fpr,
        tpr
    )


    plt.plot(

        fpr,

        tpr,

        lw=2,

        label=(

            f"Class: "
            f"{classes[i]} "
            f"(AUC = "
            f"{roc_auc:.3f})"

        )

    )


plt.plot(

    [0, 1],

    [0, 1],

    "k--",

    lw=2,

    label="Random Guess"

)


plt.title(

    "Multi-Class Receiver Operating "
    "Characteristic (ROC) Curves",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(

    "False Positive Rate (FPR)",

    fontsize=12

)


plt.ylabel(

    "True Positive Rate (TPR)",

    fontsize=12

)


plt.legend(

    loc="lower right",

    fontsize=10

)


plt.grid(

    True,

    linestyle="--",

    alpha=0.6

)


plt.tight_layout()

plt.show()


# ============================================================
# 37. FINAL SUMMARY
# ============================================================

print(
    "\n" + "=" * 80
)

print(
    "FINAL MOBILENET SUMMARY"
)

print(
    "=" * 80
)


print(

    f"Train Accuracy       : "
    f"{train_acc * 100:.2f}%"

)


print(

    f"Validation Accuracy  : "
    f"{val_acc * 100:.2f}%"

)


print(

    f"Test Accuracy        : "
    f"{test_accuracy_percentage:.2f}%"

)


print(

    f"Parameters           : "
    f"{total_params_m:.2f} M"

)


print(

    f"FLOPs                : "
    f"{gflops:.2f} GFLOPs"

)


if not np.isnan(
    peak_memory_mb
):

    print(

        f"Peak GPU Memory      : "
        f"{peak_memory_mb:.2f} MB"

    )

else:

    print(
        "Peak GPU Memory      : "
        "Unavailable"
    )


print(

    f"Inference Time       : "
    f"{avg_inference_time_ms:.4f} ms/image"

)


print(
    "=" * 80
)
```
