```python
# ============================================================
# RESNET50V2 CLASSIFICATION
# ACTUAL GPU MEMORY + FLOPs + PARAMETERS + INFERENCE TIME
# TRAIN / VALIDATION / TEST ACCURACY WITH 95% CI
# ============================================================

import os
import glob
import time
import gc
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    roc_curve,
    auc
)
from sklearn.preprocessing import LabelBinarizer

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.resnet_v2 import (
    ResNet50V2,
    preprocess_input
)
from tensorflow.keras import Model, layers
from tensorflow.keras.utils import to_categorical


# ============================================================
# 0. GPU CONFIGURATION
# ============================================================

print("=" * 80)
print("SYSTEM INFORMATION")
print("=" * 80)

gpus = tf.config.list_physical_devices("GPU")

if gpus:
    print("GPU detected:")
    for gpu in gpus:
        print("   ", gpu)

    # Enable memory growth
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except RuntimeError as e:
        print("Memory growth configuration:", e)

else:
    print("WARNING: No GPU detected. Running on CPU.")

print("TensorFlow version:", tf.__version__)
print("=" * 80)


# ============================================================
# 1. DATASET
# ============================================================

DATASET_PATH = (
    "/kaggle/input/datasets/"
    "mdhasanahmad/fresh-water-dataset/"
    "Original Image"
)

classes = sorted(
    [
        d for d in os.listdir(DATASET_PATH)
        if os.path.isdir(os.path.join(DATASET_PATH, d))
    ]
)

print("\nClasses:")
for i, c in enumerate(classes):
    print(f"{i}: {c}")

print("\nNumber of classes:", len(classes))


# ============================================================
# 2. COLLECT IMAGE PATHS
# ============================================================

image_paths = []
image_labels = []

for idx, class_name in enumerate(classes):

    class_dir = os.path.join(DATASET_PATH, class_name)

    files = []

    for ext in ["*.jpg", "*.jpeg", "*.png", "*.bmp", "*.JPG", "*.JPEG", "*.PNG"]:
        files.extend(glob.glob(os.path.join(class_dir, ext)))

    image_paths.extend(files)
    image_labels.extend([idx] * len(files))


image_paths = np.array(image_paths)
image_labels = np.array(image_labels)

print("\nTotal images:", len(image_paths))


# ============================================================
# 3. DATASET DISTRIBUTION
# ============================================================

print("\nClass distribution:")

for idx, class_name in enumerate(classes):
    count = np.sum(image_labels == idx)
    print(f"{class_name:<25}: {count}")


# ============================================================
# 4. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# 70% TRAIN
# 10% VALIDATION
# 20% TEST

train_paths, temp_paths, train_labels, temp_labels = train_test_split(
    image_paths,
    image_labels,
    test_size=0.30,
    stratify=image_labels,
    random_state=42
)

val_paths, test_paths, val_labels, test_labels = train_test_split(
    temp_paths,
    temp_labels,
    test_size=2/3,
    stratify=temp_labels,
    random_state=42
)

print("\nDataset split:")
print("Train:", len(train_paths))
print("Validation:", len(val_paths))
print("Test:", len(test_paths))

print(
    "\nPercentages:"
)
print(
    f"Train      : {len(train_paths)/len(image_paths)*100:.2f}%"
)
print(
    f"Validation : {len(val_paths)/len(image_paths)*100:.2f}%"
)
print(
    f"Test       : {len(test_paths)/len(image_paths)*100:.2f}%"
)


# ============================================================
# 5. IMAGE CONFIGURATION
# ============================================================

IMG_SIZE = (224, 224)
BATCH_SIZE = 32


# ============================================================
# 6. DATA GENERATOR
# ============================================================

def generator(
    paths,
    labels,
    batch_size=BATCH_SIZE,
    augment=False
):

    datagen = ImageDataGenerator(
        preprocessing_function=preprocess_input,

        horizontal_flip=augment,

        zoom_range=0.2 if augment else 0
    )

    while True:

        for i in range(0, len(paths), batch_size):

            batch_paths = paths[i:i + batch_size]
            batch_labels = labels[i:i + batch_size]

            imgs = []
            labs = []

            for p, l in zip(batch_paths, batch_labels):

                img = tf.keras.preprocessing.image.load_img(
                    p,
                    target_size=IMG_SIZE
                )

                img = tf.keras.preprocessing.image.img_to_array(
                    img
                )

                imgs.append(img)
                labs.append(l)

            imgs = np.array(imgs)

            imgs = datagen.standardize(imgs)

            labs = to_categorical(
                labs,
                num_classes=len(classes)
            )

            yield imgs, labs


# ============================================================
# 7. CREATE GENERATORS
# ============================================================

train_gen = generator(
    train_paths,
    train_labels,
    augment=True
)

val_gen = generator(
    val_paths,
    val_labels,
    augment=False
)

test_gen = generator(
    test_paths,
    test_labels,
    augment=False
)


# ============================================================
# 8. STEPS
# ============================================================

# IMPORTANT:
# Use ceil so that the final partial batch is also evaluated.

train_steps = int(
    np.ceil(len(train_paths) / BATCH_SIZE)
)

val_steps = int(
    np.ceil(len(val_paths) / BATCH_SIZE)
)

test_steps = int(
    np.ceil(len(test_paths) / BATCH_SIZE)
)

print("\nSteps:")
print("Train steps:", train_steps)
print("Validation steps:", val_steps)
print("Test steps:", test_steps)


# ============================================================
# 9. BUILD RESNET50V2
# ============================================================

print("\nBuilding ResNet50V2...")

base = ResNet50V2(
    include_top=False,
    weights="imagenet",
    input_shape=(224, 224, 3)
)


# Freeze backbone
for layer in base.layers:
    layer.trainable = False


# Classification head
x = base.output

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dense(
    256,
    activation="relu"
)(x)

x = layers.Dropout(0.5)(x)

out = layers.Dense(
    len(classes),
    activation="softmax"
)(x)


model = Model(
    inputs=base.input,
    outputs=out
)


# ============================================================
# 10. COMPILE MODEL
# ============================================================

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 11. MODEL SUMMARY
# ============================================================

model.summary()


# ============================================================
# 12. TRAIN MODEL
# ============================================================

print("\n" + "=" * 80)
print("STARTING TRAINING")
print("=" * 80)

history = model.fit(
    train_gen,
    steps_per_epoch=train_steps,

    validation_data=val_gen,
    validation_steps=val_steps,

    epochs=10,

    verbose=1
)


# ============================================================
# 13. TRAIN / VALIDATION ACCURACY
# ============================================================

train_acc = history.history["accuracy"][-1]
val_acc = history.history["val_accuracy"][-1]

print(
    f"\nFinal Training Accuracy: "
    f"{train_acc:.4f} ({train_acc*100:.2f}%)"
)

print(
    f"Final Validation Accuracy: "
    f"{val_acc:.4f} ({val_acc*100:.2f}%)"
)


# ============================================================
# 14. PREDICTION FUNCTION
# ============================================================

def get_predictions(gen, steps):

    preds = []
    trues = []

    for _ in range(steps):

        x, y = next(gen)

        p = model.predict(
            x,
            verbose=0
        )

        preds.append(p)
        trues.append(y)

    preds = np.vstack(preds)
    trues = np.vstack(trues)

    return (
        np.argmax(trues, axis=1),
        np.argmax(preds, axis=1),
        preds
    )


# ============================================================
# 15. GET TRAIN / VALIDATION / TEST PREDICTIONS
# ============================================================

print("\nGenerating predictions...")

y_train, y_train_pred, y_train_prob = get_predictions(
    train_gen,
    train_steps
)

y_val, y_val_pred, y_val_prob = get_predictions(
    val_gen,
    val_steps
)

y_test, y_test_pred, y_test_prob = get_predictions(
    test_gen,
    test_steps
)


# ============================================================
# 16. 95% BOOTSTRAP CONFIDENCE INTERVAL
# ============================================================

def compute_ci(
    y_true,
    y_pred,
    y_prob,
    n_bootstrap=1000
):

    rng = np.random.RandomState(42)

    acc = []
    prec = []
    rec = []
    f1 = []
    aucs = []

    n = len(y_true)

    for _ in range(n_bootstrap):

        idx = rng.randint(
            0,
            n,
            n
        )

        yt = y_true[idx]
        yp = y_pred[idx]
        ypr = y_prob[idx]

        # Need all classes for multiclass AUC
        if len(np.unique(yt)) < len(classes):
            continue

        acc.append(
            accuracy_score(
                yt,
                yp
            )
        )

        prec.append(
            precision_score(
                yt,
                yp,
                average="macro",
                zero_division=0
            )
        )

        rec.append(
            recall_score(
                yt,
                yp,
                average="macro",
                zero_division=0
            )
        )

        f1.append(
            f1_score(
                yt,
                yp,
                average="macro",
                zero_division=0
            )
        )

        try:

            aucs.append(
                roc_auc_score(
                    to_categorical(
                        yt,
                        num_classes=len(classes)
                    ),
                    ypr,
                    multi_class="ovr",
                    average="macro"
                )
            )

        except Exception:
            pass


    def ci(values):

        values = np.array(values)

        return (
            np.mean(values),
            np.percentile(values, 2.5),
            np.percentile(values, 97.5)
        )


    return (
        ci(acc),
        ci(prec),
        ci(rec),
        ci(f1),
        ci(aucs)
    )


# ============================================================
# 17. PRINT RESULTS
# ============================================================

def print_results(
    name,
    y_true,
    y_pred,
    y_prob
):

    (
        acc_ci,
        prec_ci,
        rec_ci,
        f1_ci,
        auc_ci
    ) = compute_ci(
        y_true,
        y_pred,
        y_prob
    )


    print(
        f"\n{'=' * 70}"
    )

    print(
        f"{name.upper()} RESULTS"
    )

    print(
        f"{'=' * 70}"
    )


    print(
        f"Accuracy           : "
        f"{acc_ci[0]:.4f} "
        f"(95% CI: "
        f"{acc_ci[1]:.4f} – "
        f"{acc_ci[2]:.4f})"
    )

    print(
        f"Precision (Macro)  : "
        f"{prec_ci[0]:.4f} "
        f"(95% CI: "
        f"{prec_ci[1]:.4f} – "
        f"{prec_ci[2]:.4f})"
    )

    print(
        f"Recall (Macro)     : "
        f"{rec_ci[0]:.4f} "
        f"(95% CI: "
        f"{rec_ci[1]:.4f} – "
        f"{rec_ci[2]:.4f})"
    )

    print(
        f"F1-score (Macro)   : "
        f"{f1_ci[0]:.4f} "
        f"(95% CI: "
        f"{f1_ci[1]:.4f} – "
        f"{f1_ci[2]:.4f})"
    )

    print(
        f"AUC (OVR Macro)    : "
        f"{auc_ci[0]:.4f} "
        f"(95% CI: "
        f"{auc_ci[1]:.4f} – "
        f"{auc_ci[2]:.4f})"
    )


# ============================================================
# 18. FINAL TRAIN / VALIDATION / TEST RESULTS
# ============================================================

print_results(
    "Train",
    y_train,
    y_train_pred,
    y_train_prob
)

print_results(
    "Validation",
    y_val,
    y_val_pred,
    y_val_prob
)

print_results(
    "Test",
    y_test,
    y_test_pred,
    y_test_prob
)


# ============================================================
# 19. ACTUAL GPU MEMORY MEASUREMENT
# ============================================================

def get_gpu_memory_mb():

    gpus = tf.config.list_physical_devices("GPU")

    if not gpus:
        return None

    try:

        memory_info = tf.config.experimental.get_memory_info(
            "GPU:0"
        )

        current_mb = (
            memory_info["current"]
            / (1024 ** 2)
        )

        peak_mb = (
            memory_info["peak"]
            / (1024 ** 2)
        )

        return current_mb, peak_mb

    except Exception as e:

        print(
            "TensorFlow memory API error:",
            e
        )

        return None


# ============================================================
# 20. RESET MEMORY STATISTICS
# ============================================================

if gpus:

    try:

        tf.config.experimental.reset_memory_stats(
            "GPU:0"
        )

        print(
            "\nGPU memory statistics reset successfully."
        )

    except Exception as e:

        print(
            "Could not reset GPU memory statistics:",
            e
        )


# ============================================================
# 21. CREATE A FRESH TEST GENERATOR
# ============================================================

performance_test_gen = generator(
    test_paths,
    test_labels,
    augment=False
)


# ============================================================
# 22. WARM-UP
# ============================================================

print("\nWarming up GPU...")

warmup_x, _ = next(
    performance_test_gen
)

for _ in range(5):

    _ = model(
        warmup_x,
        training=False
    )


# Force TensorFlow execution
_ = tf.reduce_sum(
    model(
        warmup_x,
        training=False
    )
)


# ============================================================
# 23. RESET PEAK MEMORY AFTER WARMUP
# ============================================================

if gpus:

    try:

        tf.config.experimental.reset_memory_stats(
            "GPU:0"
        )

    except Exception:
        pass


# ============================================================
# 24. MEASURE INFERENCE TIME
# ============================================================

inference_times = []

num_batches_for_measurement = min(
    test_steps,
    100
)

print(
    f"\nMeasuring inference time over "
    f"{num_batches_for_measurement} batches..."
)


for _ in range(num_batches_for_measurement):

    x_batch, _ = next(
        performance_test_gen
    )

    start_time = time.perf_counter()

    predictions = model(
        x_batch,
        training=False
    )

    # Force synchronization / computation
    _ = tf.reduce_sum(predictions).numpy()

    end_time = time.perf_counter()

    batch_time_ms = (
        end_time - start_time
    ) * 1000.0

    actual_batch_size = x_batch.shape[0]

    per_image_time_ms = (
        batch_time_ms /
        actual_batch_size
    )

    inference_times.append(
        per_image_time_ms
    )


avg_inference_time_ms = np.mean(
    inference_times
)


# ============================================================
# 25. ACTUAL PEAK GPU MEMORY
# ============================================================

memory_result = get_gpu_memory_mb()

if memory_result is not None:

    current_memory_mb, peak_memory_mb = (
        memory_result
    )

else:

    current_memory_mb = np.nan
    peak_memory_mb = np.nan


# ============================================================
# 26. MODEL PARAMETERS
# ============================================================

total_params = model.count_params()

total_params_m = (
    total_params / 1e6
)


# ============================================================
# 27. FLOPs CALCULATION
# ============================================================

print("\nCalculating FLOPs...")


@tf.function
def forward_pass(x):

    return model(
        x,
        training=False
    )


try:

    concrete_func = forward_pass.get_concrete_function(
        tf.TensorSpec(
            [1, 224, 224, 3],
            tf.float32
        )
    )

    frozen_func = convert_variables_to_constants_v2(
        concrete_func
    )

except Exception:

    frozen_func = None


# ------------------------------------------------------------
# Reliable TensorFlow profiler method
# ------------------------------------------------------------

try:

    from tensorflow.python.framework.convert_to_constants import (
        convert_variables_to_constants_v2
    )

    concrete_func = tf.function(
        lambda x: model(
            x,
            training=False
        )
    ).get_concrete_function(
        tf.TensorSpec(
            [1, 224, 224, 3],
            tf.float32
        )
    )

    frozen_func = convert_variables_to_constants_v2(
        concrete_func
    )

    graph_def = frozen_func.graph.as_graph_def()

    with tf.Graph().as_default() as graph:

        tf.graph_util.import_graph_def(
            graph_def,
            name=""
        )

        run_meta = tf.compat.v1.RunMetadata()

        opts = tf.compat.v1.profiler.ProfileOptionBuilder.float_operation()

        flops_profile = tf.compat.v1.profiler.profile(
            graph=graph,
            run_meta=run_meta,
            cmd="op",
            options=opts
        )

        total_flops = (
            flops_profile.total_float_ops
        )

        gflops = (
            total_flops / 1e9
        )

except Exception as e:

    print(
        "Automatic FLOPs calculation failed:",
        e
    )

    # ResNet50V2 at 224x224
    # Approximately 6.98 GFLOPs
    gflops = 6.98


# ============================================================
# 28. TEST ACCURACY
# ============================================================

test_accuracy_percentage = (
    accuracy_score(
        y_test,
        y_test_pred
    ) * 100.0
)


# ============================================================
# 29. PERFORMANCE TABLE
# ============================================================

print(
    "\n" + "=" * 125
)

print(
    f"{'Performance & Efficiency Metrics Evaluation':^125}"
)

print(
    "=" * 125
)


header = (
    f"{'Method':<24} | "
    f"{'FLOPS (GFLOPS)':<16} | "
    f"{'Parameters (M)':<16} | "
    f"{'Memory Usage (MB)':<20} | "
    f"{'Inference Time (ms)':<21} | "
    f"{'Accuracy (%)':<13}"
)

print(header)

print("-" * 125)


row = (
    f"{'ResNet50V2 (Proposed)':<24} | "
    f"{gflops:<16.2f} | "
    f"{total_params_m:<16.2f} | "
    f"{peak_memory_mb:<20.2f} | "
    f"{avg_inference_time_ms:<21.2f} | "
    f"{test_accuracy_percentage:<13.2f}"
)

print(row)

print(
    "=" * 125
)


# ============================================================
# 30. ADDITIONAL MEMORY INFORMATION
# ============================================================

print("\nGPU MEMORY DETAILS")
print("-" * 60)

if memory_result is not None:

    print(
        f"Current GPU memory : "
        f"{current_memory_mb:.2f} MB"
    )

    print(
        f"Peak GPU memory    : "
        f"{peak_memory_mb:.2f} MB"
    )

    print(
        "\nThe table reports PEAK GPU MEMORY."
    )

else:

    print(
        "GPU memory information unavailable."
    )


print(
    f"\nTotal parameters   : "
    f"{total_params:,}"
)

print(
    f"Parameters (M)     : "
    f"{total_params_m:.2f}"
)

print(
    f"FLOPs              : "
    f"{gflops:.2f} GFLOPs"
)

print(
    f"Inference time     : "
    f"{avg_inference_time_ms:.2f} ms/image"
)

print(
    f"Test accuracy      : "
    f"{test_accuracy_percentage:.2f}%"
)


# ============================================================
# 31. ACCURACY & LOSS CURVES
# ============================================================

plt.figure(figsize=(14, 5))


# Accuracy
plt.subplot(1, 2, 1)

plt.plot(
    history.history["accuracy"],
    label="Train Accuracy",
    linewidth=2
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy",
    linewidth=2
)

plt.title(
    "Model Accuracy per Epoch",
    fontsize=12,
    fontweight="bold"
)

plt.xlabel(
    "Epoch",
    fontsize=10
)

plt.ylabel(
    "Accuracy",
    fontsize=10
)

plt.legend(
    loc="lower right"
)

plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)


# Loss
plt.subplot(1, 2, 2)

plt.plot(
    history.history["loss"],
    label="Train Loss",
    linewidth=2
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss",
    linewidth=2
)

plt.title(
    "Model Loss per Epoch",
    fontsize=12,
    fontweight="bold"
)

plt.xlabel(
    "Epoch",
    fontsize=10
)

plt.ylabel(
    "Loss",
    fontsize=10
)

plt.legend(
    loc="upper right"
)

plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)

plt.tight_layout()

plt.show()


# ============================================================
# 32. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_test_pred
)

df_cm = pd.DataFrame(
    cm,
    index=classes,
    columns=classes
)


plt.figure(
    figsize=(8, 6)
)

sns.heatmap(
    df_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    cbar=True,
    xticklabels=classes,
    yticklabels=classes
)

plt.title(
    "Confusion Matrix - Test Set",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel(
    "Predicted Label",
    fontsize=12
)

plt.ylabel(
    "True Label",
    fontsize=12
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(
    rotation=0
)

plt.tight_layout()

plt.show()


# ============================================================
# 33. MULTI-CLASS ROC CURVES
# ============================================================

lb = LabelBinarizer()

y_test_binarized = lb.fit_transform(
    y_test
)


# Binary-class correction
if (
    y_test_binarized.shape[1] == 1
    and len(classes) == 2
):

    y_test_binarized = np.hstack(
        [
            1 - y_test_binarized,
            y_test_binarized
        ]
    )


plt.figure(
    figsize=(9, 7)
)


for i in range(len(classes)):

    fpr, tpr, _ = roc_curve(
        y_test_binarized[:, i],
        y_test_prob[:, i]
    )

    roc_auc = auc(
        fpr,
        tpr
    )

    plt.plot(
        fpr,
        tpr,
        lw=2,
        label=(
            f"Class: {classes[i]} "
            f"(AUC = {roc_auc:.3f})"
        )
    )


plt.plot(
    [0, 1],
    [0, 1],
    "k--",
    lw=2,
    label="Random Guess"
)

plt.xlim(
    [0.0, 1.0]
)

plt.ylim(
    [0.0, 1.05]
)

plt.xlabel(
    "False Positive Rate (FPR)",
    fontsize=12
)

plt.ylabel(
    "True Positive Rate (TPR)",
    fontsize=12
)

plt.title(
    "Multi-Class Receiver Operating Characteristic (ROC) Curves",
    fontsize=14,
    fontweight="bold"
)

plt.legend(
    loc="lower right",
    fontsize=10
)

plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)

plt.tight_layout()

plt.show()


# ============================================================
# 34. FINAL SUMMARY
# ============================================================

print(
    "\n" + "=" * 80
)

print(
    "FINAL RESNET50V2 SUMMARY"
)

print(
    "=" * 80
)

print(
    f"Train Accuracy       : "
    f"{train_acc*100:.2f}%"
)

print(
    f"Validation Accuracy  : "
    f"{val_acc*100:.2f}%"
)

print(
    f"Test Accuracy        : "
    f"{test_accuracy_percentage:.2f}%"
)

print(
    f"Parameters           : "
    f"{total_params_m:.2f} M"
)

print(
    f"FLOPs                : "
    f"{gflops:.2f} GFLOPs"
)

print(
    f"Peak GPU Memory      : "
    f"{peak_memory_mb:.2f} MB"
)

print(
    f"Inference Time       : "
    f"{avg_inference_time_ms:.2f} ms/image"
)

print(
    "=" * 80
)
```
