```python
# ============================================================
# COMPLETE ROBUST INCEPTIONV3 IMAGE CLASSIFICATION
# ============================================================
#
# FIXES:
#   1. NO tf.image.decode_image()
#   2. PIL-based image loading
#   3. Corrupted/non-image files removed before training
#   4. TIFF/TIF supported through PIL
#   5. RGB conversion
#   6. 70% Train / 10% Validation / 20% Test
#   7. Stratified split
#   8. Train / Validation / Test metrics
#   9. Bootstrap 95% CI
#  10. Classification report
#  11. Confusion matrix
#  12. ROC curves
#  13. Actual TensorFlow GPU peak memory
#  14. Inference time
#  15. Parameter count
#  16. FLOPs
#  17. Training curves
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import time
import gc
import math
import random
import warnings

import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image, ImageOps, UnidentifiedImageError

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve,
    auc
)


warnings.filterwarnings("ignore")


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)


# ============================================================
# 3. CONFIGURATION
# ============================================================

DATASET_PATH = (
    "/kaggle/input/datasets/mdhasanahmad/fresh-water-dataset/Original Image"
   
)

IMG_SIZE = (224, 224)

BATCH_SIZE = 32

EPOCHS = 2

BOOTSTRAP_SAMPLES = 1000

LEARNING_RATE = 0.001


# ============================================================
# 4. GPU CONFIGURATION
# ============================================================

print("\n" + "=" * 90)
print("GPU CONFIGURATION")
print("=" * 90)

gpus = tf.config.list_physical_devices("GPU")

if gpus:

    print(f"GPU detected: {gpus}")

    for gpu in gpus:

        try:

            tf.config.experimental.set_memory_growth(
                gpu,
                True
            )

        except RuntimeError as e:

            print(
                f"Memory growth warning: {e}"
            )

else:

    print(
        "WARNING: No GPU detected. "
        "The model will run on CPU."
    )


# ============================================================
# 5. CHECK DATASET PATH
# ============================================================

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"\nDataset directory does not exist:\n"
        f"{DATASET_PATH}"
    )


# ============================================================
# 6. DETECT CLASS DIRECTORIES
# ============================================================

print("\n" + "=" * 90)
print("DATASET INFORMATION")
print("=" * 90)


class_labels = sorted(
    [
        folder
        for folder in os.listdir(DATASET_PATH)
        if os.path.isdir(
            os.path.join(
                DATASET_PATH,
                folder
            )
        )
    ]
)


NUM_CLASSES = len(class_labels)


if NUM_CLASSES < 2:

    raise RuntimeError(
        f"Only {NUM_CLASSES} class(es) detected. "
        f"At least two classes are required."
    )


print(
    f"Dataset path      : {DATASET_PATH}"
)

print(
    f"Number of classes : {NUM_CLASSES}"
)

print(
    f"Classes           : {class_labels}"
)


# ============================================================
# 7. SCAN ALL FILES
# ============================================================
#
# IMPORTANT:
# We intentionally DO NOT filter only by extension here.
#
# Every file is sent through PIL validation.
#
# Therefore:
#   - jpg
#   - jpeg
#   - png
#   - bmp
#   - gif
#   - webp
#   - tif
#   - tiff
#   - and other PIL-readable images
#
# can be detected correctly.
#
# Non-image files are automatically rejected.
# ============================================================

print("\n" + "=" * 90)
print("SCANNING DATASET")
print("=" * 90)


candidate_paths = []

candidate_labels = []


for class_index, class_name in enumerate(class_labels):

    class_directory = os.path.join(
        DATASET_PATH,
        class_name
    )


    class_count = 0


    for root, dirs, files in os.walk(
        class_directory
    ):

        for filename in files:

            full_path = os.path.join(
                root,
                filename
            )

            candidate_paths.append(
                full_path
            )

            candidate_labels.append(
                class_index
            )

            class_count += 1


    print(
        f"{class_name:<35} : "
        f"{class_count:>6} candidate files"
    )


print(
    f"\nTotal candidate files: "
    f"{len(candidate_paths)}"
)


# ============================================================
# 8. PIL IMAGE VALIDATION
# ============================================================

print("\n" + "=" * 90)
print("VALIDATING IMAGES WITH PIL")
print("=" * 90)


valid_image_paths = []

valid_image_labels = []

invalid_files = []


for i, (image_path, label) in enumerate(
    zip(
        candidate_paths,
        candidate_labels
    )
):

    try:

        # ----------------------------------------------------
        # First validation
        # ----------------------------------------------------

        with Image.open(image_path) as img:

            img.verify()


        # ----------------------------------------------------
        # Second validation
        # Also test RGB conversion
        # ----------------------------------------------------

        with Image.open(image_path) as img:

            img = ImageOps.exif_transpose(
                img
            )

            img = img.convert(
                "RGB"
            )

            # Test resize operation too
            img = img.resize(
                IMG_SIZE,
                Image.Resampling.BILINEAR
            )


        valid_image_paths.append(
            image_path
        )

        valid_image_labels.append(
            label
        )


    except Exception as e:

        invalid_files.append(
            (
                image_path,
                str(e)
            )
        )


    # Progress indicator

    if (
        (i + 1) % 500 == 0
        or
        (i + 1) == len(candidate_paths)
    ):

        print(
            f"Validated "
            f"{i + 1}/"
            f"{len(candidate_paths)}"
        )


# Convert to arrays

valid_image_paths = np.array(
    valid_image_paths,
    dtype=str
)

valid_image_labels = np.array(
    valid_image_labels,
    dtype=np.int32
)


print("\n" + "=" * 90)
print("VALIDATION SUMMARY")
print("=" * 90)


print(
    f"Candidate files : "
    f"{len(candidate_paths)}"
)

print(
    f"Valid images    : "
    f"{len(valid_image_paths)}"
)

print(
    f"Invalid files   : "
    f"{len(invalid_files)}"
)


# ============================================================
# 9. PRINT INVALID FILES
# ============================================================

if len(invalid_files) > 0:

    print("\n" + "=" * 90)
    print("INVALID / CORRUPTED / NON-IMAGE FILES")
    print("=" * 90)


    for file_path, error_message in invalid_files:

        print(
            f"\nFile:\n{file_path}"
        )

        print(
            f"Reason:\n{error_message}"
        )

else:

    print(
        "\nNo invalid files detected."
    )


# ============================================================
# 10. CHECK VALID DATA
# ============================================================

if len(valid_image_paths) == 0:

    raise RuntimeError(
        "No valid images remain after PIL validation."
    )


# ============================================================
# 11. VALID CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 90)
print("VALID IMAGE CLASS DISTRIBUTION")
print("=" * 90)


for class_index, class_name in enumerate(
    class_labels
):

    count = int(
        np.sum(
            valid_image_labels == class_index
        )
    )

    print(
        f"{class_name:<35} : {count:>6}"
    )


# ============================================================
# 12. STRATIFIED DATA SPLIT
# ============================================================
#
# 70% TRAIN
# 10% VALIDATION
# 20% TEST
# ============================================================

print("\n" + "=" * 90)
print("CREATING STRATIFIED DATA SPLIT")
print("=" * 90)


train_paths, temp_paths, train_labels, temp_labels = (

    train_test_split(

        valid_image_paths,

        valid_image_labels,

        test_size=0.30,

        stratify=valid_image_labels,

        random_state=SEED
    )
)


val_paths, test_paths, val_labels, test_labels = (

    train_test_split(

        temp_paths,

        temp_labels,

        test_size=(2 / 3),

        stratify=temp_labels,

        random_state=SEED
    )
)


# ============================================================
# 13. SPLIT INFORMATION
# ============================================================

print(
    f"\nTraining   : {len(train_paths):>6} "
    f"({100 * len(train_paths) / len(valid_image_paths):.2f}%)"
)

print(
    f"Validation : {len(val_paths):>6} "
    f"({100 * len(val_paths) / len(valid_image_paths):.2f}%)"
)

print(
    f"Testing    : {len(test_paths):>6} "
    f"({100 * len(test_paths) / len(valid_image_paths):.2f}%)"
)


# ============================================================
# 14. PIL IMAGE LOADER
# ============================================================
#
# THIS IS THE MAIN FIX.
#
# There is NO:
#
# tf.image.decode_image()
#
# Therefore TensorFlow will never produce:
#
# "Unknown image file format"
#
# because PIL handles the file directly.
# ============================================================

def load_single_image(
    image_path
):

    try:

        # Open image

        with Image.open(image_path) as img:

            # Correct EXIF orientation

            img = ImageOps.exif_transpose(
                img
            )

            # Convert every image to RGB

            img = img.convert(
                "RGB"
            )

            # Resize

            img = img.resize(

                IMG_SIZE,

                Image.Resampling.BILINEAR
            )

            # Convert to NumPy

            image = np.asarray(
                img,
                dtype=np.float32
            )


        # Normalize

        image = image / 255.0


        return image


    except Exception as e:

        raise RuntimeError(

            f"Failed to load image:\n"
            f"{image_path}\n"
            f"Error: {e}"
        )


# ============================================================
# 15. KERAS SEQUENCE DATA GENERATOR
# ============================================================
#
# This generator completely bypasses TensorFlow image decoding.
# ============================================================

class PILImageSequence(
    tf.keras.utils.Sequence
):

    def __init__(

        self,

        paths,

        labels,

        batch_size,

        num_classes,

        shuffle=False,

        augment=False,

        seed=42
    ):

        self.paths = np.asarray(
            paths
        )

        self.labels = np.asarray(
            labels,
            dtype=np.int32
        )

        self.batch_size = int(
            batch_size
        )

        self.num_classes = int(
            num_classes
        )

        self.shuffle = shuffle

        self.augment = augment

        self.seed = seed

        self.rng = np.random.RandomState(
            seed
        )

        self.indices = np.arange(
            len(self.paths)
        )

        self.on_epoch_end()


    def __len__(self):

        return int(
            np.ceil(
                len(self.paths)
                /
                self.batch_size
            )
        )


    def __getitem__(
        self,
        index
    ):

        start = (
            index *
            self.batch_size
        )

        end = min(

            start +
            self.batch_size,

            len(self.paths)
        )


        batch_indices = self.indices[
            start:end
        ]


        batch_paths = self.paths[
            batch_indices
        ]


        batch_labels = self.labels[
            batch_indices
        ]


        # ----------------------------------------------------
        # Allocate batch
        # ----------------------------------------------------

        images = np.empty(

            (
                len(batch_paths),

                IMG_SIZE[0],

                IMG_SIZE[1],

                3
            ),

            dtype=np.float32
        )


        # ----------------------------------------------------
        # Load each image using PIL
        # ----------------------------------------------------

        for i, image_path in enumerate(
            batch_paths
        ):

            images[i] = load_single_image(
                image_path
            )


        # ----------------------------------------------------
        # OPTIONAL TRAINING AUGMENTATION
        # ----------------------------------------------------

        if self.augment:

            images = self.apply_augmentation(
                images
            )


        # ----------------------------------------------------
        # One-hot labels
        # ----------------------------------------------------

        labels = tf.keras.utils.to_categorical(

            batch_labels,

            num_classes=self.num_classes
        )


        labels = np.asarray(
            labels,
            dtype=np.float32
        )


        return images, labels


    def apply_augmentation(
        self,
        images
    ):

        augmented_images = np.empty_like(
            images
        )


        for i in range(
            len(images)
        ):

            img = images[i]


            # ------------------------------------------------
            # Random horizontal flip
            # ------------------------------------------------

            if self.rng.rand() < 0.5:

                img = np.fliplr(
                    img
                )


            # ------------------------------------------------
            # Random brightness
            # ------------------------------------------------

            if self.rng.rand() < 0.5:

                delta = self.rng.uniform(
                    -0.10,
                    0.10
                )

                img = img + delta


            # ------------------------------------------------
            # Random contrast
            # ------------------------------------------------

            if self.rng.rand() < 0.5:

                contrast = self.rng.uniform(
                    0.90,
                    1.10
                )

                mean = np.mean(
                    img,
                    axis=(0, 1),
                    keepdims=True
                )

                img = (
                    (img - mean) *
                    contrast
                ) + mean


            # Keep valid range

            img = np.clip(

                img,

                0.0,

                1.0
            )


            augmented_images[i] = img


        return augmented_images


    def on_epoch_end(self):

        if self.shuffle:

            self.rng.shuffle(
                self.indices
            )


# ============================================================
# 16. CREATE GENERATORS
# ============================================================

print("\n" + "=" * 90)
print("CREATING PIL DATA GENERATORS")
print("=" * 90)


# Training generator
# augmentation enabled

train_gen = PILImageSequence(

    train_paths,

    train_labels,

    batch_size=BATCH_SIZE,

    num_classes=NUM_CLASSES,

    shuffle=True,

    augment=True,

    seed=SEED
)


# Training evaluation generator
# NO augmentation

train_eval_gen = PILImageSequence(

    train_paths,

    train_labels,

    batch_size=BATCH_SIZE,

    num_classes=NUM_CLASSES,

    shuffle=False,

    augment=False,

    seed=SEED
)


# Validation

val_gen = PILImageSequence(

    val_paths,

    val_labels,

    batch_size=BATCH_SIZE,

    num_classes=NUM_CLASSES,

    shuffle=False,

    augment=False,

    seed=SEED
)


# Test

test_gen = PILImageSequence(

    test_paths,

    test_labels,

    batch_size=BATCH_SIZE,

    num_classes=NUM_CLASSES,

    shuffle=False,

    augment=False,

    seed=SEED
)


# ============================================================
# 17. GENERATOR INFORMATION
# ============================================================

print(
    f"Train batches      : {len(train_gen)}"
)

print(
    f"Validation batches : {len(val_gen)}"
)

print(
    f"Test batches       : {len(test_gen)}"
)


# ============================================================
# 18. TEST THE GENERATOR BEFORE TRAINING
# ============================================================
#
# IMPORTANT:
# We deliberately load one full batch BEFORE model.fit().
#
# If an image is bad, the problem is caught HERE instead of
# halfway through an epoch.
# ============================================================

print("\n" + "=" * 90)
print("TESTING IMAGE PIPELINE BEFORE TRAINING")
print("=" * 90)


try:

    sample_x, sample_y = train_gen[0]

    print(
        f"Sample image batch shape : {sample_x.shape}"
    )

    print(
        f"Sample label batch shape : {sample_y.shape}"
    )

    print(
        f"Image minimum value      : {sample_x.min():.4f}"
    )

    print(
        f"Image maximum value      : {sample_x.max():.4f}"
    )

    print(
        "\nIMAGE PIPELINE TEST PASSED."
    )

except Exception as e:

    raise RuntimeError(

        "\nIMAGE PIPELINE FAILED BEFORE TRAINING.\n"
        f"{e}"
    )


# ============================================================
# 19. BUILD INCEPTIONV3
# ============================================================

print("\n" + "=" * 90)
print("BUILDING INCEPTIONV3")
print("=" * 90)


base = tf.keras.applications.InceptionV3(

    include_top=False,

    weights="imagenet",

    input_shape=(224, 224, 3)
)


# Freeze pretrained backbone

base.trainable = False


# ============================================================
# 20. CLASSIFICATION HEAD
# ============================================================

x = base.output


x = tf.keras.layers.GlobalAveragePooling2D()(
    x
)


x = tf.keras.layers.Dense(

    256,

    activation="relu"
)(
    x
)


x = tf.keras.layers.Dropout(
    0.5
)(
    x
)


output = tf.keras.layers.Dense(

    NUM_CLASSES,

    activation="softmax"
)(
    x
)


# ============================================================
# 21. FINAL MODEL
# ============================================================

model = tf.keras.Model(

    inputs=base.input,

    outputs=output,

    name="InceptionV3_Classifier"
)


# ============================================================
# 22. COMPILE
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=LEARNING_RATE
    ),

    loss="categorical_crossentropy",

    metrics=[
        "accuracy"
    ]
)


# ============================================================
# 23. MODEL SUMMARY
# ============================================================

model.summary()


# ============================================================
# 24. PARAMETER COUNT
# ============================================================

total_params = model.count_params()

trainable_params = np.sum(

    [
        np.prod(v.shape)
        for v in model.trainable_variables
    ]
)

non_trainable_params = (

    total_params -
    trainable_params
)


print("\n" + "=" * 90)
print("MODEL PARAMETERS")
print("=" * 90)


print(
    f"Total parameters      : "
    f"{total_params:,}"
)

print(
    f"Trainable parameters  : "
    f"{trainable_params:,}"
)

print(
    f"Non-trainable         : "
    f"{non_trainable_params:,}"
)

print(
    f"Parameters (M)        : "
    f"{total_params / 1e6:.4f}"
)


# ============================================================
# 25. TRAINING
# ============================================================

print("\n" + "=" * 90)
print("STARTING TRAINING")
print("=" * 90)


training_start = time.perf_counter()


history = model.fit(

    train_gen,

    validation_data=val_gen,

    epochs=EPOCHS,

    verbose=1
)


training_end = time.perf_counter()


training_time = (

    training_end -
    training_start
)


print("\n" + "=" * 90)
print("TRAINING COMPLETED")
print("=" * 90)


print(
    f"Total training time: "
    f"{training_time:.2f} seconds"
)


# ============================================================
# 26. PREDICTION FUNCTION
# ============================================================

def get_dataset_predictions(
    dataset,
    dataset_name
):

    print(
        f"\nPredicting {dataset_name}..."
    )


    y_true = []

    y_pred = []

    y_prob = []


    for batch_index in range(
        len(dataset)
    ):

        x_batch, y_batch = dataset[
            batch_index
        ]


        predictions = model.predict_on_batch(
            x_batch
        )


        true_labels = np.argmax(

            y_batch,

            axis=1
        )


        predicted_labels = np.argmax(

            predictions,

            axis=1
        )


        y_true.extend(
            true_labels.tolist()
        )


        y_pred.extend(
            predicted_labels.tolist()
        )


        y_prob.extend(
            predictions.tolist()
        )


    return (

        np.asarray(
            y_true,
            dtype=np.int32
        ),

        np.asarray(
            y_pred,
            dtype=np.int32
        ),

        np.asarray(
            y_prob,
            dtype=np.float32
        )
    )


# ============================================================
# 27. EXTRACT PREDICTIONS
# ============================================================

print("\n" + "=" * 90)
print("EXTRACTING PREDICTIONS")
print("=" * 90)


y_train, y_train_pred, y_train_prob = (

    get_dataset_predictions(

        train_eval_gen,

        "training set"
    )
)


y_val, y_val_pred, y_val_prob = (

    get_dataset_predictions(

        val_gen,

        "validation set"
    )
)


y_test, y_test_pred, y_test_prob = (

    get_dataset_predictions(

        test_gen,

        "test set"
    )
)


# ============================================================
# 28. VERIFY PREDICTION COUNTS
# ============================================================

print("\n" + "=" * 90)
print("PREDICTION COUNT VERIFICATION")
print("=" * 90)


print(
    f"Train:"
    f" expected={len(train_paths)},"
    f" predicted={len(y_train)}"
)


print(
    f"Validation:"
    f" expected={len(val_paths)},"
    f" predicted={len(y_val)}"
)


print(
    f"Test:"
    f" expected={len(test_paths)},"
    f" predicted={len(y_test)}"
)


if len(y_train) != len(train_paths):

    raise RuntimeError(
        "Training prediction count mismatch."
    )


if len(y_val) != len(val_paths):

    raise RuntimeError(
        "Validation prediction count mismatch."
    )


if len(y_test) != len(test_paths):

    raise RuntimeError(
        "Test prediction count mismatch."
    )


# ============================================================
# 29. BOOTSTRAP CONFIDENCE INTERVAL
# ============================================================

def compute_ci(

    y_true,

    y_pred,

    y_prob,

    n_bootstrap=1000
):

    rng = np.random.RandomState(
        SEED
    )


    accuracy_values = []

    precision_values = []

    recall_values = []

    f1_values = []

    auc_values = []


    n_samples = len(
        y_true
    )


    for _ in range(
        n_bootstrap
    ):

        indices = rng.randint(

            0,

            n_samples,

            n_samples
        )


        yt = y_true[
            indices
        ]

        yp = y_pred[
            indices
        ]

        yp_prob = y_prob[
            indices
        ]


        # Accuracy

        accuracy_values.append(

            accuracy_score(
                yt,
                yp
            )
        )


        # Precision

        precision_values.append(

            precision_score(

                yt,

                yp,

                average="macro",

                zero_division=0
            )
        )


        # Recall

        recall_values.append(

            recall_score(

                yt,

                yp,

                average="macro",

                zero_division=0
            )
        )


        # F1

        f1_values.append(

            f1_score(

                yt,

                yp,

                average="macro",

                zero_division=0
            )
        )


        # AUC

        try:

            if len(
                np.unique(yt)
            ) >= 2:

                yt_onehot = tf.keras.utils.to_categorical(

                    yt,

                    num_classes=NUM_CLASSES
                )


                auc_value = roc_auc_score(

                    yt_onehot,

                    yp_prob,

                    multi_class="ovr",

                    average="macro"
                )


                auc_values.append(
                    auc_value
                )

        except Exception:

            pass


    def summarize(values):

        values = np.asarray(
            values
        )


        if len(values) == 0:

            return (
                np.nan,
                np.nan,
                np.nan
            )


        return (

            np.mean(values),

            np.percentile(
                values,
                2.5
            ),

            np.percentile(
                values,
                97.5
            )
        )


    return (

        summarize(
            accuracy_values
        ),

        summarize(
            precision_values
        ),

        summarize(
            recall_values
        ),

        summarize(
            f1_values
        ),

        summarize(
            auc_values
        )
    )


# ============================================================
# 30. PRINT RESULTS
# ============================================================

def print_results(

    dataset_name,

    y_true,

    y_pred,

    y_prob
):

    (
        accuracy_ci,
        precision_ci,
        recall_ci,
        f1_ci,
        auc_ci
    ) = compute_ci(

        y_true,

        y_pred,

        y_prob,

        n_bootstrap=BOOTSTRAP_SAMPLES
    )


    print("\n" + "=" * 90)

    print(
        f"{dataset_name.upper()} RESULTS"
    )

    print(
        "95% BOOTSTRAP CONFIDENCE INTERVAL"
    )

    print("=" * 90)


    print(

        f"Accuracy: "
        f"{accuracy_ci[0]:.4f} "
        f"(95% CI: "
        f"{accuracy_ci[1]:.4f} – "
        f"{accuracy_ci[2]:.4f})"
    )


    print(

        f"Precision (Macro): "
        f"{precision_ci[0]:.4f} "
        f"(95% CI: "
        f"{precision_ci[1]:.4f} – "
        f"{precision_ci[2]:.4f})"
    )


    print(

        f"Recall (Macro): "
        f"{recall_ci[0]:.4f} "
        f"(95% CI: "
        f"{recall_ci[1]:.4f} – "
        f"{recall_ci[2]:.4f})"
    )


    print(

        f"F1-score (Macro): "
        f"{f1_ci[0]:.4f} "
        f"(95% CI: "
        f"{f1_ci[1]:.4f} – "
        f"{f1_ci[2]:.4f})"
    )


    if not np.isnan(auc_ci[0]):

        print(

            f"AUC (OvR Macro): "
            f"{auc_ci[0]:.4f} "
            f"(95% CI: "
            f"{auc_ci[1]:.4f} – "
            f"{auc_ci[2]:.4f})"
        )

    else:

        print(
            "AUC (OvR Macro): Could not be calculated."
        )


# ============================================================
# 31. TRAIN RESULTS
# ============================================================

print_results(

    "Train",

    y_train,

    y_train_pred,

    y_train_prob
)


# ============================================================
# 32. VALIDATION RESULTS
# ============================================================

print_results(

    "Validation",

    y_val,

    y_val_pred,

    y_val_prob
)


# ============================================================
# 33. TEST RESULTS
# ============================================================

print_results(

    "Test",

    y_test,

    y_test_pred,

    y_test_prob
)


# ============================================================
# 34. TEST CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 90)
print("TEST SET CLASSIFICATION REPORT")
print("=" * 90)


print(

    classification_report(

        y_test,

        y_test_pred,

        target_names=class_labels,

        digits=4,

        zero_division=0
    )
)


# ============================================================
# 35. GPU MEMORY FUNCTIONS
# ============================================================

def get_gpu_memory():

    gpus = tf.config.list_physical_devices(
        "GPU"
    )


    if not gpus:

        return (
            np.nan,
            np.nan
        )


    try:

        memory_info = (

            tf.config.experimental.get_memory_info(

                "GPU:0"
            )
        )


        current_mb = (

            memory_info["current"]

            /

            (1024 ** 2)
        )


        peak_mb = (

            memory_info["peak"]

            /

            (1024 ** 2)
        )


        return (

            current_mb,

            peak_mb
        )


    except Exception as e:

        print(
            f"GPU memory unavailable: {e}"
        )


        return (
            np.nan,
            np.nan
        )


# ============================================================
# 36. RESET GPU MEMORY STATISTICS
# ============================================================

def reset_gpu_memory_stats():

    gpus = tf.config.list_physical_devices(
        "GPU"
    )


    if not gpus:

        return False


    try:

        tf.config.experimental.reset_memory_stats(

            "GPU:0"
        )

        return True


    except Exception as e:

        print(
            f"Could not reset GPU memory statistics: {e}"
        )

        return False


# ============================================================
# 37. PERFORMANCE EVALUATION
# ============================================================

def evaluate_performance(

    model,

    test_dataset,

    y_true_test,

    y_pred_test
):

    print("\n" + "=" * 122)

    print(
        "PERFORMANCE & EFFICIENCY EVALUATION"
    )

    print("=" * 122)


    # --------------------------------------------------------
    # PARAMETERS
    # --------------------------------------------------------

    total_params = model.count_params()


    total_params_m = (

        total_params /
        1e6
    )


    # --------------------------------------------------------
    # APPROXIMATE FLOPs
    # --------------------------------------------------------
    #
    # Standard approximate value used for InceptionV3.
    #
    # --------------------------------------------------------

    gflops = 5.72


    # --------------------------------------------------------
    # WARMUP
    # --------------------------------------------------------

    print(
        "\nPerforming inference warm-up..."
    )


    warmup_x, _ = test_dataset[0]


    for _ in range(5):

        warmup_output = model(

            warmup_x,

            training=False
        )


        _ = tf.reduce_sum(
            warmup_output
        ).numpy()


    gc.collect()


    # --------------------------------------------------------
    # RESET MEMORY
    # --------------------------------------------------------

    memory_reset = reset_gpu_memory_stats()


    if memory_reset:

        print(
            "GPU memory statistics reset."
        )

    else:

        print(
            "GPU memory statistics reset unavailable."
        )


    # --------------------------------------------------------
    # MEASURE INFERENCE
    # --------------------------------------------------------

    print(
        "Measuring inference time..."
    )


    total_images = 0

    total_time_seconds = 0.0


    for batch_index in range(
        len(test_dataset)
    ):

        x_batch, _ = test_dataset[
            batch_index
        ]


        actual_batch_size = (

            x_batch.shape[0]
        )


        start_time = (
            time.perf_counter()
        )


        predictions = model(

            x_batch,

            training=False
        )


        # Force computation

        _ = tf.reduce_sum(
            predictions
        ).numpy()


        end_time = (
            time.perf_counter()
        )


        elapsed = (

            end_time -
            start_time
        )


        total_time_seconds += elapsed

        total_images += actual_batch_size


    # --------------------------------------------------------
    # INFERENCE TIME PER IMAGE
    # --------------------------------------------------------

    if total_images > 0:

        inference_time_ms = (

            total_time_seconds *

            1000.0 /

            total_images
        )

    else:

        inference_time_ms = np.nan


    # --------------------------------------------------------
    # GPU MEMORY
    # --------------------------------------------------------

    current_memory_mb, peak_memory_mb = (
        get_gpu_memory()
    )


    # --------------------------------------------------------
    # TEST ACCURACY
    # --------------------------------------------------------

    test_accuracy = (

        accuracy_score(

            y_true_test,

            y_pred_test
        )

        * 100.0
    )


    # --------------------------------------------------------
    # PRINT
    # --------------------------------------------------------

    print("\n" + "=" * 90)

    print("RAW PERFORMANCE RESULTS")

    print("=" * 90)


    print(
        f"Total parameters       : "
        f"{total_params:,}"
    )


    print(
        f"Parameters (M)         : "
        f"{total_params_m:.4f}"
    )


    print(
        f"FLOPs (GFLOPs)         : "
        f"{gflops:.2f}"
    )


    if np.isnan(current_memory_mb):

        print(
            "Current GPU memory     : "
            "Unavailable"
        )

    else:

        print(
            f"Current GPU memory     : "
            f"{current_memory_mb:.2f} MB"
        )


    if np.isnan(peak_memory_mb):

        print(
            "Peak GPU memory        : "
            "Unavailable"
        )

    else:

        print(
            f"Peak GPU memory        : "
            f"{peak_memory_mb:.2f} MB"
        )


    print(
        f"Inference time/image   : "
        f"{inference_time_ms:.4f} ms"
    )


    print(
        f"Test accuracy          : "
        f"{test_accuracy:.2f}%"
    )


    # --------------------------------------------------------
    # IEEE STYLE TABLE
    # --------------------------------------------------------

    print("\n" + "=" * 122)

    print(
        f"{'Performance & Efficiency Metrics Evaluation':^122}"
    )

    print("=" * 122)


    header = (

        f"{'Method':<24} | "

        f"{'FLOPs (GFLOPs)':<16} | "

        f"{'Parameters (M)':<16} | "

        f"{'Memory Usage (MB)':<20} | "

        f"{'Inference Time (ms)':<21} | "

        f"{'Accuracy (%)':<13}"
    )


    print(header)

    print("-" * 122)


    if np.isnan(peak_memory_mb):

        memory_display = "N/A"

    else:

        memory_display = (
            f"{peak_memory_mb:.2f}"
        )


    row = (

        f"{'InceptionV3':<24} | "

        f"{gflops:<16.2f} | "

        f"{total_params_m:<16.2f} | "

        f"{memory_display:<20} | "

        f"{inference_time_ms:<21.4f} | "

        f"{test_accuracy:<13.2f}"
    )


    print(row)

    print("=" * 122)


    return {

        "Model":
        "InceptionV3",

        "FLOPs (GFLOPs)":
        gflops,

        "Parameters (M)":
        total_params_m,

        "Peak Memory (MB)":
        peak_memory_mb,

        "Current Memory (MB)":
        current_memory_mb,

        "Inference Time (ms/image)":
        inference_time_ms,

        "Accuracy (%)":
        test_accuracy
    }


# ============================================================
# 38. RUN PERFORMANCE EVALUATION
# ============================================================

performance_results = evaluate_performance(

    model,

    test_gen,

    y_test,

    y_test_pred
)


# ============================================================
# 39. PERFORMANCE DATAFRAME
# ============================================================

performance_df = pd.DataFrame(

    [performance_results]
)


print("\n" + "=" * 90)

print("PERFORMANCE DATAFRAME")

print("=" * 90)


print(
    performance_df.to_string(
        index=False
    )
)


# ============================================================
# 40. SAVE PERFORMANCE CSV
# ============================================================

performance_df.to_csv(

    "/kaggle/working/InceptionV3_Performance.csv",

    index=False
)


print(
    "\nSaved:"
    "/kaggle/working/InceptionV3_Performance.csv"
)


# ============================================================
# 41. TRAINING CURVES
# ============================================================

print("\nGenerating training curves...")


plt.figure(
    figsize=(14, 5)
)


# Accuracy

plt.subplot(
    1,
    2,
    1
)


plt.plot(

    history.history["accuracy"],

    label="Train Accuracy",

    linewidth=2
)


plt.plot(

    history.history["val_accuracy"],

    label="Validation Accuracy",

    linewidth=2
)


plt.title(

    "InceptionV3 Accuracy",

    fontsize=13,

    fontweight="bold"
)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)


plt.legend()

plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)


# Loss

plt.subplot(
    1,
    2,
    2
)


plt.plot(

    history.history["loss"],

    label="Train Loss",

    linewidth=2
)


plt.plot(

    history.history["val_loss"],

    label="Validation Loss",

    linewidth=2
)


plt.title(

    "InceptionV3 Loss",

    fontsize=13,

    fontweight="bold"
)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)


plt.legend()

plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)


plt.tight_layout()

plt.show()


# ============================================================
# 42. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    y_test_pred
)


plt.figure(

    figsize=(9, 7)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=class_labels,

    yticklabels=class_labels
)


plt.title(

    "Confusion Matrix - Test Set",

    fontsize=14,

    fontweight="bold"
)


plt.xlabel(
    "Predicted Label"
)


plt.ylabel(
    "True Label"
)


plt.xticks(

    rotation=45,

    ha="right"
)


plt.yticks(
    rotation=0
)


plt.tight_layout()

plt.show()


# ============================================================
# 43. ROC CURVES
# ============================================================

print("\nGenerating ROC curves...")


y_test_binarized = tf.keras.utils.to_categorical(

    y_test,

    num_classes=NUM_CLASSES
)


plt.figure(

    figsize=(10, 8)
)


for i in range(NUM_CLASSES):

    if len(
        np.unique(
            y_test_binarized[:, i]
        )
    ) < 2:

        print(
            f"Skipping ROC for "
            f"{class_labels[i]}"
        )

        continue


    fpr, tpr, _ = roc_curve(

        y_test_binarized[:, i],

        y_test_prob[:, i]
    )


    roc_auc = auc(

        fpr,

        tpr
    )


    plt.plot(

        fpr,

        tpr,

        linewidth=2,

        label=(

            f"{class_labels[i]} "

            f"(AUC = {roc_auc:.4f})"
        )
    )


# Random classifier

plt.plot(

    [0, 1],

    [0, 1],

    "k--",

    linewidth=2,

    label="Random Guess"
)


plt.xlim(
    0.0,
    1.0
)


plt.ylim(
    0.0,
    1.05
)


plt.xlabel(
    "False Positive Rate (FPR)"
)


plt.ylabel(
    "True Positive Rate (TPR)"
)


plt.title(

    "Multi-Class ROC Curves",

    fontsize=14,

    fontweight="bold"
)


plt.legend(
    loc="lower right"
)


plt.grid(

    True,

    linestyle="--",

    alpha=0.6
)


plt.tight_layout()

plt.show()


# ============================================================
# 44. SAVE MODEL
# ============================================================

MODEL_PATH = (

    "/kaggle/working/"
    "InceptionV3_final.keras"
)


model.save(
    MODEL_PATH
)


print(
    f"\nModel saved to:\n{MODEL_PATH}"
)


# ============================================================
# 45. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)

print("FINAL INCEPTIONV3 SUMMARY")

print("=" * 100)


print(
    f"Model                     : InceptionV3"
)


print(
    f"Input size                : "
    f"{IMG_SIZE[0]} × {IMG_SIZE[1]}"
)


print(
    f"Number of classes         : "
    f"{NUM_CLASSES}"
)


print(
    f"Valid images              : "
    f"{len(valid_image_paths)}"
)


print(
    f"Invalid files removed     : "
    f"{len(invalid_files)}"
)


print(
    f"Training images           : "
    f"{len(train_paths)}"
)


print(
    f"Validation images         : "
    f"{len(val_paths)}"
)


print(
    f"Testing images            : "
    f"{len(test_paths)}"
)


print(
    f"Total parameters          : "
    f"{model.count_params():,}"
)


print(
    f"Parameters (M)            : "
    f"{model.count_params() / 1e6:.4f}"
)


print(
    f"Approx. FLOPs             : "
    f"5.72 GFLOPs"
)


if np.isnan(
    performance_results[
        "Peak Memory (MB)"
    ]
):

    print(
        "Peak GPU memory          : "
        "Unavailable"
    )

else:

    print(
        f"Peak GPU memory          : "
        f"{performance_results['Peak Memory (MB)']:.2f} MB"
    )


print(
    f"Inference time/image     : "
    f"{performance_results['Inference Time (ms/image)']:.4f} ms"
)


print(
    f"Test accuracy             : "
    f"{performance_results['Accuracy (%)']:.2f}%"
)


print(
    f"Training time             : "
    f"{training_time:.2f} seconds"
)


print("=" * 100)

print(
    "EVALUATION COMPLETED SUCCESSFULLY"
)

print("=" * 100)
```
