```python
# ============================================================
# DENSENET121 - SINGLE DATASET FOLDER
# ROBUST 70% TRAIN / 10% VALIDATION / 20% TEST
# ============================================================
#
# IMPORTANT FIX:
#   NO tf.keras.preprocessing.image.load_img()
#   NO flow_from_directory()
#   NO tf.image.decode_image()
#
# Every image is:
#   1. discovered from the single dataset folder
#   2. fully decoded using PIL
#   3. converted to RGB
#   4. resized to 224x224
#   5. invalid/corrupted files removed
#   6. stratified into 70/10/20
#
# MODEL:
#   DenseNet121 ImageNet
#   GlobalAveragePooling2D
#   Dense(128, ReLU)
#   Dropout(0.5)
#   Softmax
#
# OUTPUT:
#   Train / Validation / Test metrics
#   Bootstrap 95% CI
#   Classification report
#   Confusion matrix
#   ROC curves
#   Training curves
#   Parameters
#   GFLOPs
#   Actual TensorFlow GPU peak memory
#   Inference time
#   CSV files
#   Saved model
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import time
import gc
import random
import logging
import warnings

import numpy as np
import pandas as pd
import tensorflow as tf

import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image, ImageOps, UnidentifiedImageError

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.densenet import DenseNet121, preprocess_input
from tensorflow.keras import Model, layers

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    auc
)


# ============================================================
# 2. SUPPRESS WARNINGS
# ============================================================

warnings.filterwarnings("ignore")

logger = tf.get_logger()
logger.setLevel(logging.ERROR)


# ============================================================
# 3. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)


# ============================================================
# 4. CONFIGURATION
# ============================================================

DATASET_PATH = (
    "/kaggle/input/datasets/mdhasanahmad/fresh-water-dataset/Original Image"
    
)

IMG_SIZE = (224, 224)

BATCH_SIZE = 32

EPOCHS = 10

LEARNING_RATE = 0.001

BOOTSTRAP_SAMPLES = 1000


# ============================================================
# 5. GPU CONFIGURATION
# ============================================================

print("\n" + "=" * 100)
print("GPU CONFIGURATION")
print("=" * 100)

gpus = tf.config.list_physical_devices("GPU")

if gpus:

    print("GPU detected:")

    for gpu in gpus:
        print(gpu)

        try:
            tf.config.experimental.set_memory_growth(
                gpu,
                True
            )

        except RuntimeError as e:

            print(
                f"Memory growth warning: {e}"
            )

else:

    print(
        "WARNING: No GPU detected."
    )


# ============================================================
# 6. CHECK DATASET PATH
# ============================================================

if not os.path.isdir(DATASET_PATH):

    raise FileNotFoundError(
        f"\nDataset directory does not exist:\n"
        f"{DATASET_PATH}"
    )


print(
    f"\nDataset path:\n{DATASET_PATH}"
)


# ============================================================
# 7. DETECT CLASS FOLDERS
# ============================================================

print("\n" + "=" * 100)
print("DETECTING CLASSES")
print("=" * 100)

classes = sorted(
    [
        name
        for name in os.listdir(DATASET_PATH)
        if os.path.isdir(
            os.path.join(DATASET_PATH, name)
        )
    ]
)

NUM_CLASSES = len(classes)

if NUM_CLASSES < 2:

    raise RuntimeError(
        f"Only {NUM_CLASSES} classes detected."
    )


print(
    f"Number of classes: {NUM_CLASSES}"
)

for idx, class_name in enumerate(classes):

    print(
        f"{idx}: {class_name}"
    )


# ============================================================
# 8. FULL IMAGE VALIDATION
# ============================================================
#
# CRITICAL:
#
# img.verify() alone can miss some files that fail later when
# pixels are actually decoded.
#
# Therefore this function:
#
#   verify()
#   reopen
#   img.load()
#   convert("RGB")
#   resize()
#   np.asarray()
#
# This makes the validation much stricter.
# ============================================================

def fully_validate_image(file_path):

    try:

        # ----------------------------------------------------
        # First verification
        # ----------------------------------------------------

        with Image.open(file_path) as img:

            img.verify()


        # ----------------------------------------------------
        # Re-open after verify()
        # ----------------------------------------------------

        with Image.open(file_path) as img:

            # Handle EXIF orientation

            img = ImageOps.exif_transpose(img)

            # ------------------------------------------------
            # FORCE FULL IMAGE DECODING
            # ------------------------------------------------

            img.load()

            # ------------------------------------------------
            # Convert to RGB
            # ------------------------------------------------

            img = img.convert("RGB")

            # ------------------------------------------------
            # Resize
            # ------------------------------------------------

            img = img.resize(
                IMG_SIZE,
                Image.Resampling.BILINEAR
            )

            # ------------------------------------------------
            # Force complete NumPy conversion
            # ------------------------------------------------

            arr = np.asarray(
                img,
                dtype=np.uint8
            )

            # ------------------------------------------------
            # Validate dimensions
            # ------------------------------------------------

            expected_shape = (
                IMG_SIZE[0],
                IMG_SIZE[1],
                3
            )

            if arr.shape != expected_shape:

                raise ValueError(
                    f"Unexpected image shape: "
                    f"{arr.shape}"
                )


        return True, ""


    except Exception as e:

        return False, str(e)


# ============================================================
# 9. SCAN ENTIRE SINGLE DATASET FOLDER
# ============================================================

print("\n" + "=" * 100)
print("SCANNING AND VALIDATING ALL FILES")
print("=" * 100)

image_paths = []

image_labels = []

invalid_files = []


for class_index, class_name in enumerate(classes):

    class_dir = os.path.join(
        DATASET_PATH,
        class_name
    )

    candidate_count = 0
    valid_count = 0


    for root, dirs, files in os.walk(class_dir):

        for filename in files:

            file_path = os.path.join(
                root,
                filename
            )

            candidate_count += 1


            is_valid, error_message = (
                fully_validate_image(
                    file_path
                )
            )


            if is_valid:

                image_paths.append(
                    file_path
                )

                image_labels.append(
                    class_index
                )

                valid_count += 1

            else:

                invalid_files.append(
                    (
                        file_path,
                        error_message
                    )
                )


    print(
        f"{class_name:<30} : "
        f"{valid_count:>6} valid / "
        f"{candidate_count:>6} files"
    )


# Convert to arrays

image_paths = np.asarray(
    image_paths,
    dtype=str
)

image_labels = np.asarray(
    image_labels,
    dtype=np.int32
)


# ============================================================
# 10. VALIDATION SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("IMAGE VALIDATION SUMMARY")
print("=" * 100)

print(
    f"Candidate files : "
    f"{len(image_paths) + len(invalid_files)}"
)

print(
    f"Valid images    : "
    f"{len(image_paths)}"
)

print(
    f"Invalid files   : "
    f"{len(invalid_files)}"
)


# ============================================================
# 11. PRINT INVALID FILES
# ============================================================

if len(invalid_files) > 0:

    print("\n" + "=" * 100)
    print("INVALID / CORRUPTED / NON-IMAGE FILES")
    print("=" * 100)

    for file_path, error_message in invalid_files:

        print(
            f"\nFile:\n{file_path}"
        )

        print(
            f"Reason:\n{error_message}"
        )

else:

    print(
        "\nNo invalid files detected."
    )


# ============================================================
# 12. CHECK DATA
# ============================================================

if len(image_paths) == 0:

    raise RuntimeError(
        "No valid images remain after validation."
    )


# ============================================================
# 13. CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 100)
print("VALID IMAGE DISTRIBUTION")
print("=" * 100)

for class_index, class_name in enumerate(classes):

    count = int(
        np.sum(
            image_labels == class_index
        )
    )

    print(
        f"{class_name:<30} : "
        f"{count:>6}"
    )


# ============================================================
# 14. STRATIFIED 70/10/20 SPLIT
# ============================================================
#
# FIRST:
#   70% train
#   30% temporary
#
# SECOND:
#   1/3 temporary = 10% validation
#   2/3 temporary = 20% test
# ============================================================

print("\n" + "=" * 100)
print("CREATING STRATIFIED 70/10/20 SPLIT")
print("=" * 100)


train_paths, temp_paths, train_labels, temp_labels = (

    train_test_split(

        image_paths,

        image_labels,

        test_size=0.30,

        stratify=image_labels,

        random_state=SEED
    )
)


val_paths, test_paths, val_labels, test_labels = (

    train_test_split(

        temp_paths,

        temp_labels,

        test_size=2 / 3,

        stratify=temp_labels,

        random_state=SEED
    )
)


# ============================================================
# 15. SPLIT INFORMATION
# ============================================================

total_valid_images = len(
    image_paths
)


print("\n" + "=" * 100)
print("DATASET SPLIT")
print("=" * 100)

print(
    f"Training   : "
    f"{len(train_paths):>6} "
    f"({len(train_paths) / total_valid_images * 100:.2f}%)"
)

print(
    f"Validation : "
    f"{len(val_paths):>6} "
    f"({len(val_paths) / total_valid_images * 100:.2f}%)"
)

print(
    f"Testing    : "
    f"{len(test_paths):>6} "
    f"({len(test_paths) / total_valid_images * 100:.2f}%)"
)

print(
    f"Total      : "
    f"{len(train_paths) + len(val_paths) + len(test_paths)}"
)


# ============================================================
# 16. SPLIT CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 100)
print("SPLIT CLASS DISTRIBUTION")
print("=" * 100)


for class_index, class_name in enumerate(classes):

    train_count = int(
        np.sum(
            train_labels == class_index
        )
    )

    val_count = int(
        np.sum(
            val_labels == class_index
        )
    )

    test_count = int(
        np.sum(
            test_labels == class_index
        )
    )

    print(
        f"{class_name:<25} | "
        f"Train: {train_count:<6} | "
        f"Val: {val_count:<6} | "
        f"Test: {test_count:<6}"
    )


# ============================================================
# 17. DENSENET PREPROCESSING
# ============================================================

train_datagen = ImageDataGenerator(

    preprocessing_function=preprocess_input,

    horizontal_flip=True,

    zoom_range=0.20
)


val_test_datagen = ImageDataGenerator(

    preprocessing_function=preprocess_input
)


# ============================================================
# 18. ROBUST PIL DATA SEQUENCE
# ============================================================
#
# IMPORTANT:
#
# This is NOT a Python generator using load_img().
#
# It is a Keras Sequence.
#
# Every file is read directly using PIL.
# ============================================================

class RobustDenseNetSequence(
    tf.keras.utils.Sequence
):

    def __init__(

        self,

        paths,

        labels,

        batch_size,

        datagen,

        num_classes,

        shuffle=False,

        augment=False,

        seed=42

    ):

        super().__init__()


        self.paths = np.asarray(
            paths,
            dtype=str
        )

        self.labels = np.asarray(

            labels,

            dtype=np.int32
        )

        self.batch_size = int(
            batch_size
        )

        self.datagen = datagen

        self.num_classes = int(
            num_classes
        )

        self.shuffle = shuffle

        self.augment = augment

        self.rng = np.random.RandomState(
            seed
        )

        self.indices = np.arange(
            len(self.paths)
        )


        self.on_epoch_end()


    def __len__(self):

        return int(

            np.ceil(

                len(self.paths)
                /
                self.batch_size
            )
        )


    def __getitem__(self, index):

        start = (

            index *
            self.batch_size
        )


        end = min(

            start +
            self.batch_size,

            len(self.paths)
        )


        batch_indices = self.indices[
            start:end
        ]


        batch_paths = self.paths[
            batch_indices
        ]


        batch_labels = self.labels[
            batch_indices
        ]


        # ----------------------------------------------------
        # Allocate batch
        # ----------------------------------------------------

        X = np.empty(

            (
                len(batch_paths),

                IMG_SIZE[0],

                IMG_SIZE[1],

                3
            ),

            dtype=np.float32
        )


        # ----------------------------------------------------
        # Load each image using PIL
        # ----------------------------------------------------

        for i, path in enumerate(batch_paths):

            try:

                with Image.open(path) as img:

                    # Correct orientation

                    img = ImageOps.exif_transpose(
                        img
                    )

                    # Explicitly decode pixels

                    img.load()

                    # RGB

                    img = img.convert(
                        "RGB"
                    )

                    # Resize

                    img = img.resize(

                        IMG_SIZE,

                        Image.Resampling.BILINEAR
                    )

                    # NumPy

                    image_array = np.asarray(

                        img,

                        dtype=np.float32
                    )


                # ------------------------------------------------
                # Apply augmentation only for training
                # ------------------------------------------------

                if self.augment:

                    image_array = (

                        self.datagen.random_transform(

                            image_array
                        )
                    )


                # ------------------------------------------------
                # DenseNet preprocess_input
                # ------------------------------------------------

                image_array = (

                    self.datagen.standardize(

                        image_array
                    )
                )


                X[i] = image_array


            except Exception as e:

                raise RuntimeError(

                    "\nIMAGE LOADING ERROR\n"

                    f"File:\n{path}\n\n"

                    f"Error:\n{e}"
                )


        # ----------------------------------------------------
        # One-hot labels
        # ----------------------------------------------------

        y = tf.keras.utils.to_categorical(

            batch_labels,

            num_classes=self.num_classes
        )


        y = np.asarray(

            y,

            dtype=np.float32
        )


        return X, y


    def on_epoch_end(self):

        if self.shuffle:

            self.rng.shuffle(
                self.indices
            )


# ============================================================
# 19. CREATE TRAINING GENERATOR
# ============================================================

train_gen = RobustDenseNetSequence(

    train_paths,

    train_labels,

    batch_size=BATCH_SIZE,

    datagen=train_datagen,

    num_classes=NUM_CLASSES,

    shuffle=True,

    augment=True,

    seed=SEED
)


# ============================================================
# 20. TRAIN EVALUATION GENERATOR
# ============================================================
#
# No augmentation.
#
# Used for the reported train metrics.
# ============================================================

train_eval_gen = RobustDenseNetSequence(

    train_paths,

    train_labels,

    batch_size=BATCH_SIZE,

    datagen=val_test_datagen,

    num_classes=NUM_CLASSES,

    shuffle=False,

    augment=False,

    seed=SEED
)


# ============================================================
# 21. VALIDATION GENERATOR
# ============================================================

val_gen = RobustDenseNetSequence(

    val_paths,

    val_labels,

    batch_size=BATCH_SIZE,

    datagen=val_test_datagen,

    num_classes=NUM_CLASSES,

    shuffle=False,

    augment=False,

    seed=SEED
)


# ============================================================
# 22. TEST GENERATOR
# ============================================================

test_gen = RobustDenseNetSequence(

    test_paths,

    test_labels,

    batch_size=BATCH_SIZE,

    datagen=val_test_datagen,

    num_classes=NUM_CLASSES,

    shuffle=False,

    augment=False,

    seed=SEED
)


# ============================================================
# 23. TEST IMAGE PIPELINE BEFORE TRAINING
# ============================================================
#
# THIS IS VERY IMPORTANT.
#
# A complete batch is loaded before model.fit().
#
# If this succeeds, the PIL input pipeline itself is working.
# ============================================================

print("\n" + "=" * 100)
print("TESTING IMAGE PIPELINE BEFORE TRAINING")
print("=" * 100)


try:

    sample_x, sample_y = train_gen[0]


    print(
        f"Image batch shape : "
        f"{sample_x.shape}"
    )


    print(
        f"Label batch shape : "
        f"{sample_y.shape}"
    )


    print(
        f"Image min         : "
        f"{np.min(sample_x):.4f}"
    )


    print(
        f"Image max         : "
        f"{np.max(sample_x):.4f}"
    )


    print(
        "\nIMAGE PIPELINE TEST PASSED."
    )


except Exception as e:

    raise RuntimeError(

        "\nIMAGE PIPELINE TEST FAILED.\n"

        f"{e}"
    )


# ============================================================
# 24. BUILD DENSENET121
# ============================================================

print("\n" + "=" * 100)
print("BUILDING DENSENET121")
print("=" * 100)


base = DenseNet121(

    include_top=False,

    weights="imagenet",

    input_shape=(

        IMG_SIZE[0],

        IMG_SIZE[1],

        3
    )
)


# Freeze ImageNet backbone

base.trainable = False


# ============================================================
# 25. CLASSIFICATION HEAD
# ============================================================

x = base.output


x = layers.GlobalAveragePooling2D()(x)


x = layers.Dense(

    128,

    activation="relu"
)(x)


x = layers.Dropout(
    0.5
)(x)


out = layers.Dense(

    NUM_CLASSES,

    activation="softmax"
)(x)


# ============================================================
# 26. FINAL MODEL
# ============================================================

model = Model(

    inputs=base.input,

    outputs=out,

    name="DenseNet121_Freshwater_Fish"
)


# ============================================================
# 27. COMPILE
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=LEARNING_RATE
    ),

    loss="categorical_crossentropy",

    metrics=[
        "accuracy"
    ]
)


# ============================================================
# 28. MODEL SUMMARY
# ============================================================

model.summary()


# ============================================================
# 29. PARAMETER COUNT
# ============================================================

total_params = model.count_params()


trainable_params = int(

    np.sum(

        [

            np.prod(
                variable.shape
            )

            for variable
            in model.trainable_variables

        ]
    )
)


non_trainable_params = (

    total_params -
    trainable_params
)


print("\n" + "=" * 100)
print("MODEL PARAMETERS")
print("=" * 100)


print(

    f"Total parameters     : "
    f"{total_params:,}"
)


print(

    f"Trainable parameters : "
    f"{trainable_params:,}"
)


print(

    f"Non-trainable        : "
    f"{non_trainable_params:,}"
)


print(

    f"Parameters (M)       : "
    f"{total_params / 1e6:.4f}"
)


# ============================================================
# 30. TRAIN
# ============================================================

print("\n" + "=" * 100)
print("STARTING TRAINING")
print("=" * 100)


training_start = time.perf_counter()


history = model.fit(

    train_gen,

    validation_data=val_gen,

    epochs=EPOCHS,

    verbose=1
)


training_end = time.perf_counter()


training_time = (

    training_end -
    training_start
)


print("\n" + "=" * 100)
print("TRAINING COMPLETED")
print("=" * 100)


print(

    f"Total training time: "
    f"{training_time:.2f} seconds"
)


# ============================================================
# 31. PREDICTION FUNCTION
# ============================================================

def get_predictions(

    generator,

    dataset_name

):

    print(
        f"\nPredicting {dataset_name}..."
    )


    true_labels = []

    predicted_labels = []

    probabilities = []


    for batch_index in range(

        len(generator)
    ):

        X_batch, y_batch = generator[
            batch_index
        ]


        prediction = model.predict_on_batch(

            X_batch
        )


        y_true_batch = np.argmax(

            y_batch,

            axis=1
        )


        y_pred_batch = np.argmax(

            prediction,

            axis=1
        )


        true_labels.extend(

            y_true_batch.tolist()
        )


        predicted_labels.extend(

            y_pred_batch.tolist()
        )


        probabilities.extend(

            prediction.tolist()
        )


    return (

        np.asarray(
            true_labels,
            dtype=np.int32
        ),

        np.asarray(
            predicted_labels,
            dtype=np.int32
        ),

        np.asarray(
            probabilities,
            dtype=np.float32
        )
    )


# ============================================================
# 32. GET TRAIN / VALIDATION / TEST PREDICTIONS
# ============================================================

print("\n" + "=" * 100)
print("EXTRACTING PREDICTIONS")
print("=" * 100)


y_train, y_train_pred, y_train_prob = (

    get_predictions(

        train_eval_gen,

        "TRAIN"
    )
)


y_val, y_val_pred, y_val_prob = (

    get_predictions(

        val_gen,

        "VALIDATION"
    )
)


y_test, y_test_pred, y_test_prob = (

    get_predictions(

        test_gen,

        "TEST"
    )
)


# ============================================================
# 33. VERIFY COUNTS
# ============================================================

print("\n" + "=" * 100)
print("PREDICTION COUNT VERIFICATION")
print("=" * 100)


print(

    f"Train:"
    f" expected={len(train_paths)},"
    f" predicted={len(y_train)}"
)


print(

    f"Validation:"
    f" expected={len(val_paths)},"
    f" predicted={len(y_val)}"
)


print(

    f"Test:"
    f" expected={len(test_paths)},"
    f" predicted={len(y_test)}"
)


if len(y_train) != len(train_paths):

    raise RuntimeError(
        "Training prediction count mismatch."
    )


if len(y_val) != len(val_paths):

    raise RuntimeError(
        "Validation prediction count mismatch."
    )


if len(y_test) != len(test_paths):

    raise RuntimeError(
        "Test prediction count mismatch."
    )


# ============================================================
# 34. METRICS FUNCTION
# ============================================================

def compute_metrics(

    y_true,

    y_pred,

    y_prob
):

    accuracy = accuracy_score(

        y_true,

        y_pred
    )


    precision = precision_score(

        y_true,

        y_pred,

        average="macro",

        zero_division=0
    )


    recall = recall_score(

        y_true,

        y_pred,

        average="macro",

        zero_division=0
    )


    f1 = f1_score(

        y_true,

        y_pred,

        average="macro",

        zero_division=0
    )


    try:

        y_onehot = tf.keras.utils.to_categorical(

            y_true,

            num_classes=NUM_CLASSES
        )


        auc_value = roc_auc_score(

            y_onehot,

            y_prob,

            multi_class="ovr",

            average="macro"
        )

    except Exception:

        auc_value = np.nan


    return (

        accuracy,

        precision,

        recall,

        f1,

        auc_value
    )


# ============================================================
# 35. BOOTSTRAP CI
# ============================================================

def compute_bootstrap_ci(

    y_true,

    y_pred,

    y_prob,

    n_bootstrap=1000
):

    rng = np.random.RandomState(
        SEED
    )


    metric_values = []


    n = len(
        y_true
    )


    for _ in range(
        n_bootstrap
    ):

        indices = rng.randint(

            0,

            n,

            n
        )


        sampled_true = y_true[
            indices
        ]


        sampled_pred = y_pred[
            indices
        ]


        sampled_prob = y_prob[
            indices
        ]


        sampled_metrics = compute_metrics(

            sampled_true,

            sampled_pred,

            sampled_prob
        )


        metric_values.append(
            sampled_metrics
        )


    metric_values = np.asarray(

        metric_values,

        dtype=float
    )


    lower = np.nanpercentile(

        metric_values,

        2.5,

        axis=0
    )


    upper = np.nanpercentile(

        metric_values,

        97.5,

        axis=0
    )


    return lower, upper


# ============================================================
# 36. PRINT RESULTS
# ============================================================

def print_results(

    name,

    y_true,

    y_pred,

    y_prob

):

    metrics = compute_metrics(

        y_true,

        y_pred,

        y_prob
    )


    lower, upper = (

        compute_bootstrap_ci(

            y_true,

            y_pred,

            y_prob,

            n_bootstrap=BOOTSTRAP_SAMPLES
        )
    )


    print("\n" + "=" * 100)

    print(
        f"{name.upper()} RESULTS"
    )

    print(
        "95% BOOTSTRAP CONFIDENCE INTERVAL"
    )

    print("=" * 100)


    print(

        f"Accuracy: "
        f"{metrics[0]:.4f} "
        f"(95% CI: "
        f"{lower[0]:.4f} – "
        f"{upper[0]:.4f})"
    )


    print(

        f"Precision (Macro): "
        f"{metrics[1]:.4f} "
        f"(95% CI: "
        f"{lower[1]:.4f} – "
        f"{upper[1]:.4f})"
    )


    print(

        f"Recall (Macro): "
        f"{metrics[2]:.4f} "
        f"(95% CI: "
        f"{lower[2]:.4f} – "
        f"{upper[2]:.4f})"
    )


    print(

        f"F1-score (Macro): "
        f"{metrics[3]:.4f} "
        f"(95% CI: "
        f"{lower[3]:.4f} – "
        f"{upper[3]:.4f})"
    )


    if np.isnan(metrics[4]):

        print(
            "AUC (OvR Macro): Unavailable"
        )

    else:

        print(

            f"AUC (OvR Macro): "
            f"{metrics[4]:.4f} "
            f"(95% CI: "
            f"{lower[4]:.4f} – "
            f"{upper[4]:.4f})"
        )


    return {

        "Accuracy": metrics[0],

        "Precision": metrics[1],

        "Recall": metrics[2],

        "F1": metrics[3],

        "AUC": metrics[4],

        "Accuracy CI Lower": lower[0],

        "Accuracy CI Upper": upper[0],

        "Precision CI Lower": lower[1],

        "Precision CI Upper": upper[1],

        "Recall CI Lower": lower[2],

        "Recall CI Upper": upper[2],

        "F1 CI Lower": lower[3],

        "F1 CI Upper": upper[3],

        "AUC CI Lower": lower[4],

        "AUC CI Upper": upper[4]
    }


# ============================================================
# 37. TRAIN RESULTS
# ============================================================

train_results = print_results(

    "Train",

    y_train,

    y_train_pred,

    y_train_prob
)


# ============================================================
# 38. VALIDATION RESULTS
# ============================================================

validation_results = print_results(

    "Validation",

    y_val,

    y_val_pred,

    y_val_prob
)


# ============================================================
# 39. TEST RESULTS
# ============================================================

test_results = print_results(

    "Test",

    y_test,

    y_test_pred,

    y_test_prob
)


# ============================================================
# 40. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 100)

print("TEST SET CLASSIFICATION REPORT")

print("=" * 100)


print(

    classification_report(

        y_test,

        y_test_pred,

        target_names=classes,

        digits=4,

        zero_division=0
    )
)


# ============================================================
# 41. GPU MEMORY FUNCTIONS
# ============================================================

def get_gpu_memory():

    gpus = tf.config.list_physical_devices(
        "GPU"
    )


    if not gpus:

        return (
            np.nan,
            np.nan
        )


    try:

        memory_info = (

            tf.config.experimental.get_memory_info(

                "GPU:0"
            )
        )


        current_mb = (

            memory_info["current"]

            /

            (1024 ** 2)
        )


        peak_mb = (

            memory_info["peak"]

            /

            (1024 ** 2)
        )


        return (

            current_mb,

            peak_mb
        )


    except Exception as e:

        print(
            f"GPU memory measurement error: {e}"
        )


        return (
            np.nan,
            np.nan
        )


# ============================================================
# 42. RESET GPU MEMORY
# ============================================================

def reset_gpu_memory_stats():

    gpus = tf.config.list_physical_devices(
        "GPU"
    )


    if not gpus:

        return False


    try:

        tf.config.experimental.reset_memory_stats(

            "GPU:0"
        )

        return True


    except Exception as e:

        print(
            f"Could not reset GPU memory: {e}"
        )

        return False


# ============================================================
# 43. PERFORMANCE EVALUATION
# ============================================================

def evaluate_tensorflow_performance(

    model,

    test_generator,

    y_true_test,

    y_pred_test
):

    print("\n" + "=" * 122)

    print(
        "PERFORMANCE & EFFICIENCY EVALUATION"
    )

    print("=" * 122)


    # --------------------------------------------------------
    # PARAMETERS
    # --------------------------------------------------------

    total_params = model.count_params()


    total_params_m = (

        total_params /
        1e6
    )


    # --------------------------------------------------------
    # APPROXIMATE DENSENET121 FLOPs
    # --------------------------------------------------------

    gflops = 2.89


    # --------------------------------------------------------
    # WARMUP
    # --------------------------------------------------------

    print(
        "\nPerforming inference warm-up..."
    )


    warmup_x, _ = test_generator[0]


    for _ in range(5):

        warmup_output = model(

            warmup_x,

            training=False
        )


        _ = tf.reduce_sum(

            warmup_output
        ).numpy()


    gc.collect()


    # --------------------------------------------------------
    # RESET MEMORY STATISTICS
    # --------------------------------------------------------

    memory_reset = (

        reset_gpu_memory_stats()
    )


    if memory_reset:

        print(
            "GPU peak-memory counter reset."
        )

    else:

        print(
            "GPU peak-memory counter could not be reset."
        )


    # --------------------------------------------------------
    # INFERENCE TIME
    # --------------------------------------------------------

    print(
        "Measuring inference time..."
    )


    total_images = 0

    total_time_seconds = 0.0


    for batch_index in range(

        len(test_generator)
    ):

        X_batch, _ = test_generator[
            batch_index
        ]


        current_batch_size = int(

            X_batch.shape[0]
        )


        start_time = (

            time.perf_counter()
        )


        output = model(

            X_batch,

            training=False
        )


        # Force computation

        _ = tf.reduce_sum(

            output
        ).numpy()


        end_time = (

            time.perf_counter()
        )


        elapsed = (

            end_time -
            start_time
        )


        total_time_seconds += elapsed

        total_images += current_batch_size


    # --------------------------------------------------------
    # TIME PER IMAGE
    # --------------------------------------------------------

    if total_images > 0:

        inference_time_ms = (

            total_time_seconds *

            1000.0 /

            total_images
        )

    else:

        inference_time_ms = np.nan


    # --------------------------------------------------------
    # GPU MEMORY
    # --------------------------------------------------------

    current_memory_mb, peak_memory_mb = (

        get_gpu_memory()
    )


    # --------------------------------------------------------
    # TEST ACCURACY
    # --------------------------------------------------------

    test_accuracy_percentage = (

        accuracy_score(

            y_true_test,

            y_pred_test
        )

        *

        100.0
    )


    # --------------------------------------------------------
    # RAW RESULTS
    # --------------------------------------------------------

    print("\n" + "=" * 90)

    print(
        "RAW PERFORMANCE MEASUREMENTS"
    )

    print("=" * 90)


    print(

        f"Total parameters       : "
        f"{total_params:,}"
    )


    print(

        f"Parameters (M)         : "
        f"{total_params_m:.4f}"
    )


    print(

        f"FLOPs (GFLOPs)         : "
        f"{gflops:.2f}"
    )


    if np.isnan(current_memory_mb):

        print(
            "Current GPU memory     : "
            "Unavailable"
        )

    else:

        print(

            f"Current GPU memory     : "
            f"{current_memory_mb:.2f} MB"
        )


    if np.isnan(peak_memory_mb):

        print(
            "Peak GPU memory        : "
            "Unavailable"
        )

    else:

        print(

            f"Peak GPU memory        : "
            f"{peak_memory_mb:.2f} MB"
        )


    print(

        f"Inference time/image   : "
        f"{inference_time_ms:.4f} ms"
    )


    print(

        f"Test accuracy          : "
        f"{test_accuracy_percentage:.2f}%"
    )


    # ========================================================
    # IEEE TABLE
    # ========================================================

    print("\n" + "=" * 122)

    print(

        f"{'Performance & Efficiency Metrics Evaluation':^122}"
    )

    print("=" * 122)


    print(

        f"{'Method':<24} | "
        f"{'FLOPs (GFLOPS)':<16} | "
        f"{'Parameters (M)':<16} | "
        f"{'Memory Usage (MB)':<20} | "
        f"{'Inference Time (ms)':<21} | "
        f"{'Accuracy (%)':<13}"
    )


    print("-" * 122)


    if np.isnan(
        peak_memory_mb
    ):

        memory_display = "N/A"

    else:

        memory_display = (
            f"{peak_memory_mb:.2f}"
        )


    print(

        f"{'DenseNet121':<24} | "

        f"{gflops:<16.2f} | "

        f"{total_params_m:<16.2f} | "

        f"{memory_display:<20} | "

        f"{inference_time_ms:<21.4f} | "

        f"{test_accuracy_percentage:<13.2f}"
    )


    print("=" * 122)


    return {

        "Model":
        "DenseNet121",

        "FLOPs (GFLOPs)":
        gflops,

        "Parameters (M)":
        total_params_m,

        "Peak Memory (MB)":
        peak_memory_mb,

        "Current Memory (MB)":
        current_memory_mb,

        "Inference Time (ms/image)":
        inference_time_ms,

        "Accuracy (%)":
        test_accuracy_percentage
    }


# ============================================================
# 44. RUN PERFORMANCE EVALUATION
# ============================================================

performance_results = (

    evaluate_tensorflow_performance(

        model,

        test_gen,

        y_test,

        y_test_pred
    )
)


# ============================================================
# 45. PERFORMANCE DATAFRAME
# ============================================================

performance_df = pd.DataFrame(

    [performance_results]
)


print("\n" + "=" * 100)

print("PERFORMANCE DATAFRAME")

print("=" * 100)


print(

    performance_df.to_string(
        index=False
    )
)


# ============================================================
# 46. SAVE PERFORMANCE CSV
# ============================================================

performance_csv = (

    "/kaggle/working/"
    "DenseNet121_Freshwater_Performance.csv"
)


performance_df.to_csv(

    performance_csv,

    index=False
)


print(
    f"\nPerformance CSV saved to:\n"
    f"{performance_csv}"
)


# ============================================================
# 47. TRAINING CURVES
# ============================================================

plt.figure(

    figsize=(14, 5)
)


# ------------------------------------------------------------
# ACCURACY
# ------------------------------------------------------------

plt.subplot(

    1,

    2,

    1
)


plt.plot(

    history.history["accuracy"],

    label="Train Accuracy",

    linewidth=2
)


plt.plot(

    history.history["val_accuracy"],

    label="Validation Accuracy",

    linewidth=2
)


plt.title(

    "DenseNet121 Model Accuracy",

    fontsize=13,

    fontweight="bold"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.legend(

    loc="lower right"
)


plt.grid(

    True,

    linestyle="--",

    alpha=0.6
)


# ------------------------------------------------------------
# LOSS
# ------------------------------------------------------------

plt.subplot(

    1,

    2,

    2
)


plt.plot(

    history.history["loss"],

    label="Train Loss",

    linewidth=2
)


plt.plot(

    history.history["val_loss"],

    label="Validation Loss",

    linewidth=2
)


plt.title(

    "DenseNet121 Model Loss",

    fontsize=13,

    fontweight="bold"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Loss"
)


plt.legend(

    loc="upper right"
)


plt.grid(

    True,

    linestyle="--",

    alpha=0.6
)


plt.tight_layout()

plt.show()


# ============================================================
# 48. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    y_test_pred,

    labels=np.arange(
        NUM_CLASSES
    )
)


print("\n" + "=" * 100)

print("CONFUSION MATRIX - TEST SET")

print("=" * 100)


plt.figure(

    figsize=(13, 11)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=classes,

    yticklabels=classes
)


plt.title(

    "DenseNet121 Confusion Matrix - Test Set",

    fontsize=14,

    fontweight="bold"
)


plt.xlabel(
    "Predicted Label"
)


plt.ylabel(
    "True Label"
)


plt.xticks(

    rotation=45,

    ha="right"
)


plt.yticks(
    rotation=0
)


plt.tight_layout()

plt.show()


# ============================================================
# 49. ROC CURVES
# ============================================================

print("\n" + "=" * 100)

print("GENERATING MULTI-CLASS ROC CURVES")

print("=" * 100)


y_test_onehot = tf.keras.utils.to_categorical(

    y_test,

    num_classes=NUM_CLASSES
)


plt.figure(

    figsize=(11, 9)
)


for class_index in range(
    NUM_CLASSES
):

    class_true = (

        y_test_onehot[
            :,
            class_index
        ]
    )


    class_prob = (

        y_test_prob[
            :,
            class_index
        ]
    )


    if len(
        np.unique(class_true)
    ) < 2:

        print(

            f"Skipping ROC for "
            f"{classes[class_index]}"
        )

        continue


    fpr, tpr, _ = roc_curve(

        class_true,

        class_prob
    )


    class_auc = auc(

        fpr,

        tpr
    )


    plt.plot(

        fpr,

        tpr,

        linewidth=2,

        label=(

            f"{classes[class_index]} "
            f"(AUC = {class_auc:.3f})"
        )
    )


# Random guess line

plt.plot(

    [0, 1],

    [0, 1],

    "k--",

    linewidth=2,

    label="Random Guess"
)


plt.xlim(
    0.0,
    1.0
)


plt.ylim(
    0.0,
    1.05
)


plt.xlabel(

    "False Positive Rate (FPR)",

    fontsize=12
)


plt.ylabel(

    "True Positive Rate (TPR)",

    fontsize=12
)


plt.title(

    "DenseNet121 Multi-Class ROC Curves",

    fontsize=14,

    fontweight="bold"
)


plt.legend(

    loc="lower right",

    fontsize=8
)


plt.grid(

    True,

    linestyle="--",

    alpha=0.6
)


plt.tight_layout()

plt.show()


# ============================================================
# 50. SAVE MODEL
# ============================================================

model_path = (

    "/kaggle/working/"
    "DenseNet121_Freshwater_Final.keras"
)


model.save(
    model_path
)


print(
    f"\nModel saved to:\n"
    f"{model_path}"
)


# ============================================================
# 51. SAVE TEST PREDICTIONS
# ============================================================

prediction_df = pd.DataFrame({

    "True_Label":
        y_test,

    "Predicted_Label":
        y_test_pred,

    "True_Class": [

        classes[index]

        for index in y_test
    ],

    "Predicted_Class": [

        classes[index]

        for index in y_test_pred
    ]
})


prediction_csv = (

    "/kaggle/working/"
    "DenseNet121_Freshwater_Test_Predictions.csv"
)


prediction_df.to_csv(

    prediction_csv,

    index=False
)


print(

    f"Test predictions saved to:\n"
    f"{prediction_csv}"
)


# ============================================================
# 52. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 110)

print("FINAL DENSENET121 SUMMARY")

print("=" * 110)


print(

    f"Dataset folder          : "
    f"{DATASET_PATH}"
)


print(

    f"Number of classes       : "
    f"{NUM_CLASSES}"
)


print(

    f"Input size              : "
    f"{IMG_SIZE[0]} × {IMG_SIZE[1]}"
)


print(

    f"Valid images            : "
    f"{len(image_paths)}"
)


print(

    f"Invalid files removed   : "
    f"{len(invalid_files)}"
)


print(

    f"Training images         : "
    f"{len(train_paths)}"
)


print(

    f"Validation images       : "
    f"{len(val_paths)}"
)


print(

    f"Testing images          : "
    f"{len(test_paths)}"
)


print(

    f"Total parameters        : "
    f"{model.count_params():,}"
)


print(

    f"Parameters (M)          : "
    f"{model.count_params() / 1e6:.4f}"
)


print(
    "Approx. FLOPs           : "
    "2.89 GFLOPs"
)


print(

    f"Test accuracy           : "
    f"{performance_results['Accuracy (%)']:.2f}%"
)


if np.isnan(

    performance_results[
        "Peak Memory (MB)"
    ]

):

    print(
        "Peak GPU memory         : "
        "Unavailable"
    )

else:

    print(

        f"Peak GPU memory         : "
        f"{performance_results['Peak Memory (MB)']:.2f} MB"
    )


print(

    f"Inference time/image    : "
    f"{performance_results['Inference Time (ms/image)']:.4f} ms"
)


print(

    f"Training time           : "
    f"{training_time:.2f} seconds"
)


print("\nSaved files:")

print(
    model_path
)

print(
    performance_csv
)

print(
    prediction_csv
)


print("\n" + "=" * 110)

print(
    "DENSENET121 EXPERIMENT COMPLETED SUCCESSFULLY"
)

print("=" * 110)
```
